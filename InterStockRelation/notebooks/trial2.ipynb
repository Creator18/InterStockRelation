```python
# Cell 1 â€” Imports and Setup
import os
import warnings
import numpy as np
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns

warnings.filterwarnings('ignore')

# Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', lambda x: f'{x:.4f}')

print("All imports successful.")
print(f"Pandas version:  {pd.__version__}")
print(f"NumPy version:   {np.__version__}")
```

```python
# Cell 2 â€” Load Raw Data from Local Archive

import os
import glob

# â”€â”€ Paths â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
PROJECT_DIR  = r"C:\Users\HP\Documents\InterStockRelation_Paper\InterStockRelation"
ARCHIVE_DIR  = os.path.join(PROJECT_DIR, "archive")

os.chdir(PROJECT_DIR)
print(f"Project directory : {PROJECT_DIR}")
print(f"Archive directory : {ARCHIVE_DIR}")

# â”€â”€ Separate index/VIX files from individual stock files â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
EXCLUDE = {"INDIA VIX_5minute.csv", "NIFTY 50_5minute.csv", "NIFTY BANK_5minute.csv"}

all_files   = sorted(glob.glob(os.path.join(ARCHIVE_DIR, "*.csv")))
index_files = [f for f in all_files if os.path.basename(f) in EXCLUDE]
stock_files = [f for f in all_files if os.path.basename(f) not in EXCLUDE]

print(f"\nTotal CSV files found : {len(all_files)}")
print(f"Index / VIX files     : {len(index_files)}  â†’ kept separately as market context")
print(f"Individual stock files: {len(stock_files)}")
print(f"Possible stock pairs  : {len(stock_files) * (len(stock_files) - 1) // 2}")

# â”€â”€ Load one file to inspect columns â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
sample_path = stock_files[0]
sample_name = os.path.basename(sample_path).replace("_5minute.csv", "")
sample_df   = pd.read_csv(sample_path)

print(f"\nSample file   : {sample_name}")
print(f"Shape         : {sample_df.shape}")
print(f"Columns       : {list(sample_df.columns)}")
print(f"\nFirst 3 rows:")
print(sample_df.head(3))
print(f"\nLast 3 rows:")
print(sample_df.tail(3))
print(f"\nDtypes:")
print(sample_df.dtypes)
print(f"\nNull counts:")
print(sample_df.isnull().sum())
```

```python
# Cell 3 â€” Load All Stocks and Preprocess

# â”€â”€ Load all stock CSVs into a dictionary â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
print("Loading all stock files...")

stocks = {}
failed = []

for fpath in stock_files:
    ticker = os.path.basename(fpath).replace("_5minute.csv", "")
    try:
        df = pd.read_csv(fpath, parse_dates=["date"])
        df = df.sort_values("date").reset_index(drop=True)
        df.set_index("date", inplace=True)
        stocks[ticker] = df
    except Exception as e:
        failed.append((ticker, str(e)))

print(f"âœ“ Successfully loaded : {len(stocks)} stocks")
if failed:
    print(f"âœ— Failed to load     : {len(failed)} stocks â†’ {failed}")

# â”€â”€ Compute log returns for each stock â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
print("\nComputing log returns...")

for ticker, df in stocks.items():
    df["log_return"] = np.log(df["close"] / df["close"].shift(1))

print("âœ“ Log returns computed for all stocks.")

# â”€â”€ Sanity check â€” confirm all stocks share a common date range â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
start_dates = pd.Series({t: df.index.min() for t, df in stocks.items()})
end_dates   = pd.Series({t: df.index.max() for t, df in stocks.items()})

print(f"\nDate range across all stocks:")
print(f"  Earliest start : {start_dates.min()}  â†’ {start_dates.idxmin()}")
print(f"  Latest start   : {start_dates.max()}  â†’ {start_dates.idxmax()}")
print(f"  Earliest end   : {end_dates.min()}    â†’ {end_dates.idxmin()}")
print(f"  Latest end     : {end_dates.max()}    â†’ {end_dates.idxmax()}")

# â”€â”€ Row count distribution â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
row_counts = pd.Series({t: len(df) for t, df in stocks.items()})
print(f"\nRow count stats across all stocks:")
print(f"  Min  : {row_counts.min()}  â†’ {row_counts.idxmin()}")
print(f"  Max  : {row_counts.max()}  â†’ {row_counts.idxmax()}")
print(f"  Mean : {row_counts.mean():.0f}")
print(f"  Std  : {row_counts.std():.0f}")

# â”€â”€ Load index and VIX files separately â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
print("\nLoading index / VIX files...")

indices = {}
for fpath in index_files:
    name = os.path.basename(fpath).replace("_5minute.csv", "")
    df   = pd.read_csv(fpath, parse_dates=["date"])
    df   = df.sort_values("date").reset_index(drop=True)
    df.set_index("date", inplace=True)
    df["log_return"] = np.log(df["close"] / df["close"].shift(1))
    indices[name] = df
    print(f"  âœ“ {name}  shape: {df.shape}")

print("\nAll data loaded and ready.")
print(f"\nstocks  dict â†’ {len(stocks)} entries")
print(f"indices dict â†’ {len(indices)} entries")
```

```python
# Cell 4 â€” Preprocessing: Filter, Align, and Prepare (with VIX)

PROJECT_DIR = r"C:\Users\HP\Documents\InterStockRelation_Paper\InterStockRelation"

# â”€â”€ Minimum row threshold â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
MIN_ROWS = 18900 * 3

row_counts      = pd.Series({t: len(df) for t, df in stocks.items()})
valid_tickers   = row_counts[row_counts >= MIN_ROWS].index.tolist()
dropped_tickers = row_counts[row_counts <  MIN_ROWS].index.tolist()

print(f"Minimum row threshold : {MIN_ROWS:,} bars (~3 years)")
print(f"âœ“ Valid stocks        : {len(valid_tickers)}")
print(f"âœ— Dropped stocks      : {len(dropped_tickers)} â†’ {dropped_tickers}")

stocks = {t: stocks[t] for t in valid_tickers}

# â”€â”€ Filter to market hours â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
print("\nFiltering to market hours (09:15â€“15:30)...")
for ticker, df in stocks.items():
    mask = (df.index.time >= pd.Timestamp("09:15").time()) & \
           (df.index.time <= pd.Timestamp("15:30").time())
    stocks[ticker] = df[mask]

# â”€â”€ Build common timestamp index â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
print("Building common timestamp index...")
common_index = None
for ticker, df in stocks.items():
    common_index = df.index if common_index is None \
                   else common_index.intersection(df.index)

print(f"âœ“ Common timestamps : {len(common_index):,}")
print(f"  From : {common_index.min()}")
print(f"  To   : {common_index.max()}")

# â”€â”€ Align stocks â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
print("\nAligning stocks...")
aligned = {}
for ticker, df in stocks.items():
    aligned[ticker] = df.loc[common_index, ["close", "log_return", "volume"]].copy()

shapes = set(df.shape for df in aligned.values())
print(f"âœ“ All aligned. Unique shapes: {shapes}")

# â”€â”€ Core matrices â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
close_matrix  = pd.DataFrame({t: aligned[t]["close"]      for t in aligned})
return_matrix = pd.DataFrame({t: aligned[t]["log_return"]  for t in aligned})
volume_matrix = pd.DataFrame({t: aligned[t]["volume"]      for t in aligned})

print(f"\nClose matrix  : {close_matrix.shape}")
print(f"Return matrix : {return_matrix.shape}")
print(f"Volume matrix : {volume_matrix.shape}")

# â”€â”€ Align VIX and NIFTY 50 to common index â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
print("\nAligning VIX and NIFTY 50...")

vix = indices["INDIA VIX"][["close"]].copy()
vix = vix.reindex(common_index).ffill().bfill()
vix.columns = ["vix"]
vix["vix_norm"] = (vix["vix"] - vix["vix"].mean()) / vix["vix"].std()

nifty = indices["NIFTY 50"][["close"]].copy()
nifty = nifty.reindex(common_index).ffill().bfill()
nifty["log_return"] = np.log(nifty["close"] / nifty["close"].shift(1)).fillna(0)
nifty.columns       = ["nifty_close", "nifty_return"]

# â”€â”€ Market context matrix â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
# This will be added as extra features in the relationship dataset
market_context = pd.DataFrame({
    "vix"         : vix["vix"],
    "vix_norm"    : vix["vix_norm"],
    "nifty_return": nifty["nifty_return"],
}, index=common_index)

print(f"âœ“ VIX shape           : {vix.shape}")
print(f"âœ“ NIFTY 50 shape      : {nifty.shape}")
print(f"âœ“ Market context shape: {market_context.shape}")
print(f"\nMarket context sample:")
print(market_context.head(3))
print(f"\nVIX stats:")
print(vix["vix"].describe().round(2))

print(f"\nâœ“ Preprocessing complete. Ready for R(t) computation.")
```

```python
# Cell 5 â€” R(t) Computation: Memory-Efficient Rewrite
# Strategy: pre-allocate numpy arrays, write directly, no per-pair DataFrames
# Uses float32 (half memory of float64, sufficient for [-100,+100] scores)

from itertools import combinations
import time

# â”€â”€ Try to use GPU (cupy), fall back to numpy silently â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
try:
    import cupy as cp
    xp = cp
    print("âœ“ CuPy found â€” using GPU (RTX 4070)")
except ImportError:
    xp = np
    print("â†’ CuPy not available â€” using CPU (numpy)")

# â”€â”€ Window / EWMA configuration â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
WINDOW_SINGLE = 75
WINDOW_SHORT  = 12
WINDOW_MEDIUM = 156
WINDOW_LONG   = 375
EWMA_ALPHA    = 1 - 0.94

print(f"\nWindows â†’ single:{WINDOW_SINGLE} | short:{WINDOW_SHORT} | "
      f"medium:{WINDOW_MEDIUM} | long:{WINDOW_LONG} | ewma_alpha:{EWMA_ALPHA:.4f}")

# â”€â”€ Pre-compute return matrix as float32 numpy array â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
tickers    = list(return_matrix.columns)
R_mat      = return_matrix.values.astype(np.float32)   # shape (T, N)
T, N       = R_mat.shape
all_pairs  = list(combinations(range(N), 2))
n_pairs    = len(all_pairs)
print(f"\nReturn matrix : {R_mat.shape}  dtype={R_mat.dtype}")
print(f"Pairs         : {n_pairs}")

# â”€â”€ Helper functions (numpy-based, operate on 1D arrays) â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
def rolling_corr(a, b, w):
    """Rolling Pearson correlation Ã— 100, output float32."""
    s = pd.Series(a)
    t = pd.Series(b)
    return (s.rolling(w).corr(t) * 100).values.astype(np.float32)

def ewma_corr(a, b, alpha):
    """EWMA correlation Ã— 100, output float32."""
    s  = pd.Series(a)
    t  = pd.Series(b)
    m1 = s.ewm(alpha=alpha, adjust=False).mean()
    m2 = t.ewm(alpha=alpha, adjust=False).mean()
    cov  = ((s - m1) * (t - m2)).ewm(alpha=alpha, adjust=False).mean()
    var1 = ((s - m1) ** 2).ewm(alpha=alpha, adjust=False).mean()
    var2 = ((t - m2) ** 2).ewm(alpha=alpha, adjust=False).mean()
    corr = cov / np.sqrt(var1 * var2)
    return (corr * 100).values.astype(np.float32)

def velocity(arr):
    """First difference (velocity), float32."""
    out      = np.empty_like(arr)
    out[0]   = np.nan
    out[1:]  = arr[1:] - arr[:-1]
    return out

def acceleration(v_arr):
    """Second difference (acceleration), float32."""
    return velocity(v_arr)

# â”€â”€ Pre-allocate output arrays  â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
# Columns per pair: R_single, R_single_v, R_single_a,
#                   R_short, R_medium, R_long, R_long_v, R_long_a,
#                   R_ewma, R_ewma_v, R_ewma_a  â†’ 11 columns
N_COLS = 11
print(f"\nPre-allocating output array: ({n_pairs}, {T}, {N_COLS}) float32")
estimated_gb = (n_pairs * T * N_COLS * 4) / (1024 ** 3)
print(f"Estimated memory            : {estimated_gb:.2f} GB")

# We store as (n_pairs Ã— T Ã— N_COLS) but write pair-by-pair to avoid peak spike
# Instead write directly to a memory-mapped file on disk
OUTPUT_PATH = os.path.join(PROJECT_DIR, "relationship_raw.npy")
output_mmap = np.lib.format.open_memmap(
    OUTPUT_PATH,
    mode  = "w+",
    dtype = np.float32,
    shape = (n_pairs, T, N_COLS)
)
print(f"âœ“ Memory-mapped file created : {OUTPUT_PATH}")

# Also store pair index metadata
pair_meta = [(tickers[i], tickers[j]) for i, j in all_pairs]

# â”€â”€ Main computation loop â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
print(f"\nComputing R(t) for {n_pairs} pairs...")
start_time = time.time()

for idx, (i, j) in enumerate(all_pairs):

    if idx % 500 == 0:
        elapsed = time.time() - start_time
        rate    = idx / elapsed if elapsed > 0 else 0
        eta     = (n_pairs - idx) / rate if rate > 0 else 0
        print(f"  {idx:>5}/{n_pairs} | {elapsed:>6.0f}s elapsed | ETA {eta:>6.0f}s")

    a = R_mat[:, i]
    b = R_mat[:, j]

    r_s  = rolling_corr(a, b, WINDOW_SINGLE)
    r_sh = rolling_corr(a, b, WINDOW_SHORT)
    r_m  = rolling_corr(a, b, WINDOW_MEDIUM)
    r_l  = rolling_corr(a, b, WINDOW_LONG)
    r_e  = ewma_corr(a, b, EWMA_ALPHA)

    r_s_v  = velocity(r_s);     r_s_a  = acceleration(r_s_v)
    r_l_v  = velocity(r_l);     r_l_a  = acceleration(r_l_v)
    r_e_v  = velocity(r_e);     r_e_a  = acceleration(r_e_v)

    output_mmap[idx, :, 0]  = r_s
    output_mmap[idx, :, 1]  = r_s_v
    output_mmap[idx, :, 2]  = r_s_a
    output_mmap[idx, :, 3]  = r_sh
    output_mmap[idx, :, 4]  = r_m
    output_mmap[idx, :, 5]  = r_l
    output_mmap[idx, :, 6]  = r_l_v
    output_mmap[idx, :, 7]  = r_l_a
    output_mmap[idx, :, 8]  = r_e
    output_mmap[idx, :, 9]  = r_e_v
    output_mmap[idx, :, 10] = r_e_a

# Flush to disk
output_mmap.flush()
elapsed_total = time.time() - start_time
print(f"\nâœ“ Computation complete in {elapsed_total:.0f}s  ({elapsed_total/60:.1f} min)")
print(f"âœ“ Saved to: {OUTPUT_PATH}")
print(f"  Shape : {output_mmap.shape}  dtype : {output_mmap.dtype}")

# Save pair metadata
import pickle
META_PATH = os.path.join(PROJECT_DIR, "pair_meta.pkl")
with open(META_PATH, "wb") as f:
    pickle.dump({"pairs": pair_meta, "tickers": tickers,
                 "timestamps": common_index, "columns": [
                     "R_single","R_single_v","R_single_a",
                     "R_short","R_medium","R_long","R_long_v","R_long_a",
                     "R_ewma","R_ewma_v","R_ewma_a"
                 ]}, f)
print(f"âœ“ Pair metadata saved : {META_PATH}")
```

```python
# Cell 8 â€” Temporal Stability Analysis

import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import numpy as np
import pandas as pd
import pickle
import os

PROJECT_DIR = r"C:\Users\HP\Documents\InterStockRelation_Paper\InterStockRelation"
OUTPUT_PATH = os.path.join(PROJECT_DIR, "relationship_raw.npy")
META_PATH   = os.path.join(PROJECT_DIR, "pair_meta.pkl")

with open(META_PATH, "rb") as f:
    meta = pickle.load(f)

pair_meta  = meta["pairs"]
col_names  = meta["columns"]
timestamps = meta["timestamps"]

mmap = np.lib.format.open_memmap(OUTPUT_PATH, mode="r")

# â”€â”€ Helper: get pair data as DataFrame â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
def get_pair_df(ticker_a, ticker_b):
    if (ticker_a, ticker_b) in pair_meta:
        idx = pair_meta.index((ticker_a, ticker_b))
    elif (ticker_b, ticker_a) in pair_meta:
        idx = pair_meta.index((ticker_b, ticker_a))
    else:
        return None, None
    df = pd.DataFrame(mmap[idx], index=timestamps, columns=col_names)
    return df, idx

# â”€â”€ Define pairs to plot â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
# 1. Same sector â€” high expected correlation
# 2. Same sector â€” moderate expected correlation
# 3. Cross sector â€” lower expected correlation
# 4. Defensive vs cyclical â€” potentially negative at times

pairs_to_plot = [
    ("HDFCBANK",  "ICICIBANK",  "Same Sector â€” Large Private Banks"),
    ("RELIANCE",  "ONGC",       "Same Sector â€” Energy"),
    ("RELIANCE",  "INFY",       "Cross Sector â€” Energy vs IT"),
    ("BRITANNIA", "COALINDIA",  "Cross Sector â€” FMCG vs Mining"),
]

# â”€â”€ Downsample for plotting â€” use daily close of R(t) â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
# 75 bars/day â†’ take every 75th row as end-of-day snapshot
EOD_STEP = 75

fig, axes = plt.subplots(4, 1, figsize=(16, 20), sharex=False)
fig.suptitle("R(t) Temporal Evolution â€” Selected Pairs (Nifty 100, 5-min)",
             fontsize=14, fontweight="bold")

for ax, (ta, tb, label) in zip(axes, pairs_to_plot):
    df, idx = get_pair_df(ta, tb)
    if df is None:
        ax.set_title(f"{ta} vs {tb} â€” NOT FOUND")
        continue

    # End-of-day snapshots
    df_eod = df.iloc[::EOD_STEP].dropna()

    ax.plot(df_eod.index, df_eod["R_single"], color="steelblue",
            lw=0.8, alpha=0.7, label="Single (1d)")
    ax.plot(df_eod.index, df_eod["R_long"],   color="forestgreen",
            lw=1.2, alpha=0.9, label="Long (1w)")
    ax.plot(df_eod.index, df_eod["R_ewma"],   color="crimson",
            lw=0.8, alpha=0.6, label="EWMA")

    # Shade positive/negative regions
    ax.axhline(0,   color="black", lw=1.0, linestyle="--", alpha=0.5)
    ax.axhline(50,  color="green", lw=0.6, linestyle=":",  alpha=0.4)
    ax.axhline(-50, color="red",   lw=0.6, linestyle=":",  alpha=0.4)
    ax.fill_between(df_eod.index, df_eod["R_long"], 0,
                    where=(df_eod["R_long"] > 0),
                    alpha=0.08, color="green")
    ax.fill_between(df_eod.index, df_eod["R_long"], 0,
                    where=(df_eod["R_long"] < 0),
                    alpha=0.08, color="red")

    ax.set_title(f"{ta}  vs  {tb}  â€”  {label}", fontsize=11, fontweight="bold")
    ax.set_ylabel("R(t)")
    ax.set_ylim(-105, 105)
    ax.legend(fontsize=8, loc="upper right")
    ax.grid(True, alpha=0.2)
    ax.xaxis.set_major_formatter(mdates.DateFormatter("%Y-%m"))
    ax.xaxis.set_major_locator(mdates.MonthLocator(interval=6))
    plt.setp(ax.xaxis.get_majorticklabels(), rotation=30, ha="right")

plt.tight_layout()
PLOT3_PATH = os.path.join(PROJECT_DIR, "plot3_temporal.png")
plt.savefig(PLOT3_PATH, dpi=150, bbox_inches="tight")
plt.show()
print(f"âœ“ Saved: {PLOT3_PATH}")

# â”€â”€ Print regime shift stats for HDFCBANK/ICICIBANK â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
df_banks, _ = get_pair_df("HDFCBANK", "ICICIBANK")
df_banks_eod = df_banks.iloc[::EOD_STEP].dropna()

# Count how many times R_long crosses zero (relationship sign flips)
sign_changes = (np.diff(np.sign(df_banks_eod["R_long"])) != 0).sum()
# Count how many times R_long drops more than 20 points in one week
big_drops = (df_banks_eod["R_long"].diff() < -20).sum()

print(f"\nâ”€â”€ HDFCBANK vs ICICIBANK â€” Relationship Dynamics â”€â”€")
print(f"  R_long sign flips (zero crossings) : {sign_changes}")
print(f"  R_long drops > 20pts in one step   : {big_drops}")
print(f"  R_long mean                        : {df_banks_eod['R_long'].mean():.2f}")
print(f"  R_long std                         : {df_banks_eod['R_long'].std():.2f}")

# Same for RELIANCE/INFY
df_rel, _ = get_pair_df("RELIANCE", "INFY")
df_rel_eod = df_rel.iloc[::EOD_STEP].dropna()
sign_changes2 = (np.diff(np.sign(df_rel_eod["R_long"])) != 0).sum()
big_drops2    = (df_rel_eod["R_long"].diff() < -20).sum()

print(f"\nâ”€â”€ RELIANCE vs INFY â€” Relationship Dynamics â”€â”€")
print(f"  R_long sign flips (zero crossings) : {sign_changes2}")
print(f"  R_long drops > 20pts in one step   : {big_drops2}")
print(f"  R_long mean                        : {df_rel_eod['R_long'].mean():.2f}")
print(f"  R_long std                         : {df_rel_eod['R_long'].std():.2f}")

print(f"\nâœ“ Cell 8 complete.")
```

```python
# Cell 9 â€” Method Comparison: Noise vs Responsiveness

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import pickle, os

PROJECT_DIR = r"C:\Users\HP\Documents\InterStockRelation_Paper\InterStockRelation"
mmap      = np.lib.format.open_memmap(os.path.join(PROJECT_DIR, "relationship_raw.npy"), mode="r")
with open(os.path.join(PROJECT_DIR, "pair_meta.pkl"), "rb") as f:
    meta = pickle.load(f)
pair_meta  = meta["pairs"]
col_names  = meta["columns"]
timestamps = meta["timestamps"]

# â”€â”€ 1. Noise level: std of velocity across all pairs (sampled) â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
# Lower std of velocity = smoother signal = less noise
SAMPLE = slice(None, None, 20)  # every 20th pair

v_single = mmap[SAMPLE, :, 1].flatten();  v_single = v_single[~np.isnan(v_single)]
v_long   = mmap[SAMPLE, :, 6].flatten();  v_long   = v_long[~np.isnan(v_long)]
v_ewma   = mmap[SAMPLE, :, 9].flatten();  v_ewma   = v_ewma[~np.isnan(v_ewma)]

print("â”€â”€ Noise Level (std of R_velocity â€” lower = smoother) â”€â”€")
print(f"  Single : {v_single.std():.4f}")
print(f"  Long   : {v_long.std():.4f}")
print(f"  EWMA   : {v_ewma.std():.4f}")

# â”€â”€ 2. Responsiveness: how fast each method reacts to a known shock â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
# Use HDFCBANK/ICICIBANK â€” find the biggest R_long drop and measure
# how many bars earlier Single and EWMA detected the same drop

def get_pair_df(ta, tb):
    key = (ta, tb) if (ta, tb) in pair_meta else (tb, ta)
    idx = pair_meta.index(key)
    return pd.DataFrame(mmap[idx], index=timestamps, columns=col_names)

df = get_pair_df("HDFCBANK", "ICICIBANK").dropna()

# Find top 5 largest single-step drops in R_long
top_drops = df["R_long_v"].nsmallest(5)
print(f"\nâ”€â”€ Top 5 Largest R_long Drops (HDFCBANK/ICICIBANK) â”€â”€")
print(top_drops.to_string())

# For each drop event, check if Single/EWMA dropped N bars earlier
print(f"\nâ”€â”€ Lead time: did Single/EWMA drop before R_long? â”€â”€")
print(f"{'Event':<30} {'Single_v at t':>15} {'EWMA_v at t':>13} {'Single_v t-5':>14} {'EWMA_v t-5':>12}")
print("-" * 88)
for ts, val in top_drops.items():
    pos = df.index.get_loc(ts)
    if pos < 5:
        continue
    sv_now  = df["R_single_v"].iloc[pos]
    ev_now  = df["R_ewma_v"].iloc[pos]
    sv_pre  = df["R_single_v"].iloc[pos - 5]
    ev_pre  = df["R_ewma_v"].iloc[pos - 5]
    print(f"{str(ts):<30} {sv_now:>15.2f} {ev_now:>13.2f} {sv_pre:>14.2f} {ev_pre:>12.2f}")

# â”€â”€ 3. Visual comparison on one turbulent period â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
# Zoom into the biggest drop event
biggest_drop_ts = top_drops.idxmin()
pos = df.index.get_loc(biggest_drop_ts)
window = slice(max(0, pos - 375), min(len(df), pos + 375))  # Â±1 week around event
df_zoom = df.iloc[window]

fig, axes = plt.subplots(3, 1, figsize=(14, 10), sharex=True)
fig.suptitle(f"Method Comparison Around Biggest Drop Event\n"
             f"HDFCBANK vs ICICIBANK  |  Event: {biggest_drop_ts}", fontsize=12)

for ax, (col, color, label) in zip(axes, [
    ("R_single", "steelblue",  "Single (1d)"),
    ("R_long",   "forestgreen","Long (1w)"),
    ("R_ewma",   "crimson",    "EWMA (Î»=0.94)"),
]):
    ax.plot(df_zoom.index, df_zoom[col], color=color, lw=1.2, label=label)
    ax.axvline(biggest_drop_ts, color="black", lw=1.5, linestyle="--", alpha=0.7, label="Drop event")
    ax.axhline(0, color="gray", lw=0.8, linestyle=":")
    ax.set_ylabel("R(t)"); ax.legend(fontsize=9); ax.grid(True, alpha=0.25)

plt.tight_layout()
path = os.path.join(PROJECT_DIR, "plot4_method_comparison.png")
plt.savefig(path, dpi=150, bbox_inches="tight")
plt.show()
print(f"\nâœ“ Saved: {path}")
print("âœ“ Cell 9 complete.")
```

```python
# Cell 11 â€” Correlation Matrix Video Construction
#
# Builds a sequence of 95Ã—95 correlation matrices â€” one per daily snapshot â€”
# where each cell (i,j) is the R_ewma between stock i and stock j.
# Stocks are ordered by the Fiedler vector so correlated stocks are spatially
# adjacent and clusters appear as bright diagonal blocks.
#
# Output: a (N_SNAPS, 95, 95) tensor â€” a "video" of market correlation structure
# Each frame is a grayscale image. Sequences of L frames are the training input.
#
# Analogy:
#   Frame  = snapshot of correlation structure at one trading day
#   Pixel  = R_ewma between two stocks (-1 to +1, like pixel intensity)
#   Bright = strongly correlated pair
#   Dark   = uncorrelated or negatively correlated pair
#   Motion = how correlation structure evolves over time

import numpy as np
import pickle, os, time
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from scipy.sparse import csr_matrix
from scipy.sparse.linalg import eigsh

PROJECT_DIR = r"C:\Users\HP\Documents\InterStockRelation_Paper\InterStockRelation"
CACHE_DIR   = os.path.join(PROJECT_DIR, "pair_cache")

with open(os.path.join(PROJECT_DIR, "pair_meta.pkl"), "rb") as f:
    meta = pickle.load(f)

pair_meta  = meta["pairs"]
tickers    = meta["tickers"]
timestamps = meta["timestamps"]
col_names  = meta["columns"]
COL        = {name: idx for idx, name in enumerate(col_names)}
ticker_idx = {t: i for i, t in enumerate(tickers)}

N_STOCKS   = len(tickers)          # 95
N_PAIRS    = len(pair_meta)        # 4465
T          = len(timestamps)       # 87122
R_EWMA_IDX = COL["R_ewma"]

STRIDE     = 75     # 1 snapshot per trading day (75 bars Ã— 5min = 375min = 1 day)
WARMUP     = 375    # skip first week while EWMA stabilises

pair_row = np.array([ticker_idx[a] for a, b in pair_meta], dtype=np.int32)
pair_col = np.array([ticker_idx[b] for a, b in pair_meta], dtype=np.int32)

print(f"Stocks     : {N_STOCKS}")
print(f"Pairs      : {N_PAIRS}")
print(f"Timesteps  : {T}")
print(f"Stride     : {STRIDE} bars (1 trading day)")

# â”€â”€ Load R_ewma for all pairs into RAM â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
print("\nLoading R_ewma into RAM...")
t0 = time.time()
r_ewma_all = np.empty((N_PAIRS, T), dtype=np.float32)
for i in range(N_PAIRS):
    arr = np.load(os.path.join(CACHE_DIR, f"pair_{i:05d}.npy"))
    r_ewma_all[i, :] = arr[:, R_EWMA_IDX]
print(f"âœ“ Loaded in {time.time()-t0:.0f}s  shape={r_ewma_all.shape}")

# â”€â”€ Compute snapshot timesteps â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
snap_timesteps = np.arange(WARMUP, T, STRIDE)
N_SNAPS        = len(snap_timesteps)
snap_dates     = [timestamps[t] for t in snap_timesteps]
print(f"\nSnapshots  : {N_SNAPS}")
print(f"Date range : {snap_dates[0]} â†’ {snap_dates[-1]}")

# â”€â”€ Compute Fiedler vector for stock ordering â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
# The Fiedler vector is the second eigenvector of the graph Laplacian.
# It gives the optimal 1D ordering of nodes so that strongly connected
# nodes are spatially adjacent â€” exactly what we need for CNN kernels
# to detect cluster blocks along the diagonal.
#
# We compute it on the MEAN correlation matrix across all snapshots
# so the ordering is stable and fixed across all frames.
# This means the relative positions of stocks never change between frames â€”
# just like objects don't teleport between video frames.

print("\nComputing mean correlation matrix for Fiedler ordering...")
t0       = time.time()
mean_adj = np.zeros((N_STOCKS, N_STOCKS), dtype=np.float32)

for snap_i, t_idx in enumerate(snap_timesteps):
    r_vals = np.nan_to_num(r_ewma_all[:, t_idx], nan=0.0) / 100.0
    # Only positive correlations contribute to the graph structure
    r_pos  = np.clip(r_vals, 0, None)
    mean_adj[pair_row, pair_col] += r_pos
    mean_adj[pair_col, pair_row] += r_pos

mean_adj /= N_SNAPS
print(f"âœ“ Mean adj computed in {time.time()-t0:.0f}s")
print(f"  Mean positive weight : {mean_adj[mean_adj > 0].mean():.4f}")

# Compute graph Laplacian: L = D - A
degree  = mean_adj.sum(axis=1)
D       = np.diag(degree)
L       = D - mean_adj

# Convert to sparse for efficient eigenvector computation
L_sparse = csr_matrix(L)

print("Computing Fiedler vector (second eigenvector of Laplacian)...")
t0 = time.time()
# k=2 gives us the two smallest eigenvalues
# eigenvalues[0] â‰ˆ 0 (trivial), eigenvalues[1] = Fiedler value
eigenvalues, eigenvectors = eigsh(L_sparse, k=2, which="SM")
fiedler_vector = eigenvectors[:, 1]   # second eigenvector
print(f"âœ“ Computed in {time.time()-t0:.1f}s")
print(f"  Fiedler value (algebraic connectivity) : {eigenvalues[1]:.6f}")
print(f"  Higher = more connected graph")

# Sort stocks by Fiedler vector value
fiedler_order  = np.argsort(fiedler_vector)
ordered_tickers= [tickers[i] for i in fiedler_order]
inverse_order  = np.argsort(fiedler_order)   # to map back to original indices

print(f"\nStock ordering by Fiedler vector:")
print(f"  First 10 : {ordered_tickers[:10]}")
print(f"  Last 10  : {ordered_tickers[-10:]}")

# â”€â”€ Build correlation matrix video â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
# corr_video[snap_i, i, j] = R_ewma between ordered_stock_i and ordered_stock_j
# Values in [-1, +1] â€” like pixel intensity in a grayscale image
# Shape: (N_SNAPS, N_STOCKS, N_STOCKS)

print(f"\nBuilding correlation matrix video...")
print(f"Shape: ({N_SNAPS}, {N_STOCKS}, {N_STOCKS})  "
      f"~{N_SNAPS * N_STOCKS * N_STOCKS * 4 / 1e6:.0f}MB")

t0          = time.time()
corr_video  = np.zeros((N_SNAPS, N_STOCKS, N_STOCKS), dtype=np.float32)

for snap_i, t_idx in enumerate(snap_timesteps):
    if snap_i % 200 == 0:
        elapsed = time.time() - t0
        eta     = (N_SNAPS - snap_i) / (snap_i / elapsed) if snap_i > 0 else 0
        print(f"  {snap_i:>5}/{N_SNAPS} | {elapsed:>5.1f}s | ETA {eta:>5.1f}s")

    # Build full correlation matrix at this snapshot
    r_vals  = np.nan_to_num(r_ewma_all[:, t_idx], nan=0.0) / 100.0
    mat     = np.zeros((N_STOCKS, N_STOCKS), dtype=np.float32)
    mat[pair_row, pair_col] = r_vals
    mat[pair_col, pair_row] = r_vals
    np.fill_diagonal(mat, 1.0)   # self-correlation = 1

    # Reorder by Fiedler vector â€” correlated stocks become spatially adjacent
    mat_ordered             = mat[np.ix_(fiedler_order, fiedler_order)]
    corr_video[snap_i, :, :]= mat_ordered

elapsed_total = time.time() - t0
print(f"\nâœ“ Video built in {elapsed_total:.0f}s ({elapsed_total/60:.1f} min)")
print(f"  Shape  : {corr_video.shape}")
print(f"  dtype  : {corr_video.dtype}")
print(f"  Min    : {corr_video.min():.4f}")
print(f"  Max    : {corr_video.max():.4f}")
print(f"  Mean   : {corr_video.mean():.4f}")

# â”€â”€ Save outputs â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
np.save(os.path.join(PROJECT_DIR, "corr_video.npy"),      corr_video)
np.save(os.path.join(PROJECT_DIR, "fiedler_order.npy"),   fiedler_order)
np.save(os.path.join(PROJECT_DIR, "snap_timesteps.npy"),  snap_timesteps)

video_meta = {
    "snap_dates"    : snap_dates,
    "snap_timesteps": snap_timesteps,
    "fiedler_order" : fiedler_order,
    "ordered_tickers": ordered_tickers,
    "tickers"       : tickers,
    "ticker_idx"    : ticker_idx,
    "n_stocks"      : N_STOCKS,
    "n_snaps"       : N_SNAPS,
    "stride"        : STRIDE,
    "warmup"        : WARMUP,
    "fiedler_value" : float(eigenvalues[1]),
}
with open(os.path.join(PROJECT_DIR, "video_meta.pkl"), "wb") as f:
    pickle.dump(video_meta, f)

print(f"\nâœ“ Saved corr_video.npy       shape={corr_video.shape}")
print(f"âœ“ Saved fiedler_order.npy    shape={fiedler_order.shape}")
print(f"âœ“ Saved snap_timesteps.npy   shape={snap_timesteps.shape}")
print(f"âœ“ Saved video_meta.pkl")

# â”€â”€ Sanity check: visualize 6 frames â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
# Should see bright diagonal blocks (clusters) that shift over time
print("\nGenerating frame visualizations...")

sample_snaps = np.linspace(0, N_SNAPS - 1, 6, dtype=int)
fig, axes    = plt.subplots(2, 3, figsize=(16, 10))
fig.suptitle("Correlation Matrix Video â€” Sample Frames\n"
             "(Stocks ordered by Fiedler vector â€” clusters appear as diagonal blocks)",
             fontsize=13, fontweight="bold")

for ax, snap_i in zip(axes.flatten(), sample_snaps):
    frame = corr_video[snap_i]
    im    = ax.imshow(frame, cmap="RdYlGn", vmin=-1, vmax=1,
                      interpolation="nearest", aspect="auto")
    ax.set_title(f"{str(snap_dates[snap_i])[:10]}", fontsize=10)
    ax.set_xlabel("Stock (Fiedler order)")
    ax.set_ylabel("Stock (Fiedler order)")
    plt.colorbar(im, ax=ax, shrink=0.8, label="R_ewma")

plt.tight_layout()
path = os.path.join(PROJECT_DIR, "plot_cell11_corr_frames.png")
plt.savefig(path, dpi=150, bbox_inches="tight")
plt.show()
print(f"âœ“ Saved: {path}")

# â”€â”€ Sanity check: pixel intensity over time for a known pair â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
# HDFCBANK and ICICIBANK should show consistently bright pixel
# RELIANCE and INFY should show dimmer, more variable pixel

print("\nPixel intensity over time for selected pairs:")
fig2, axes2 = plt.subplots(2, 1, figsize=(14, 6), sharex=True)
fig2.suptitle("Pixel Intensity Over Time â€” Selected Pairs",
              fontsize=12, fontweight="bold")

check_pairs = [
    ("HDFCBANK", "ICICIBANK", "Same cluster â€” Private Banks", "steelblue"),
    ("RELIANCE", "INFY",      "Cross cluster â€” Energy vs IT", "crimson"),
]

for ax, (ta, tb, label, color) in zip(axes2, check_pairs):
    if ta in ticker_idx and tb in ticker_idx:
        oi = inverse_order[ticker_idx[ta]]
        oj = inverse_order[ticker_idx[tb]]
        pixel_series = corr_video[:, oi, oj]
        ax.plot(snap_dates, pixel_series, lw=0.8, color=color, alpha=0.85)
        ax.axhline(0, color="black", lw=0.8, linestyle="--", alpha=0.4)
        ax.axhline(pixel_series.mean(), color="red", lw=1.2, linestyle="--",
                   alpha=0.6, label=f"Mean={pixel_series.mean():.3f}")
        ax.set_ylabel("R_ewma (pixel intensity)")
        ax.set_title(f"{ta} vs {tb} â€” {label}", fontsize=10)
        ax.set_ylim(-1, 1)
        ax.legend(fontsize=9); ax.grid(True, alpha=0.3)
    else:
        ax.set_title(f"{ta} or {tb} not found")

axes2[1].xaxis.set_major_formatter(mdates.DateFormatter("%Y-%m"))
axes2[1].xaxis.set_major_locator(mdates.MonthLocator(interval=6))
plt.setp(axes2[1].xaxis.get_majorticklabels(), rotation=30, ha="right")
plt.tight_layout()
path2 = os.path.join(PROJECT_DIR, "plot_cell11_pixel_intensity.png")
plt.savefig(path2, dpi=150, bbox_inches="tight")
plt.show()
print(f"âœ“ Saved: {path2}")

# â”€â”€ Frame statistics over time â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
frame_mean = corr_video.mean(axis=(1, 2))
frame_std  = corr_video.std(axis=(1, 2))

fig3, (ax1, ax2) = plt.subplots(2, 1, figsize=(14, 6), sharex=True)
fig3.suptitle("Frame Statistics Over Time", fontsize=12, fontweight="bold")

ax1.plot(snap_dates, frame_mean, lw=0.8, color="steelblue", alpha=0.85)
ax1.set_ylabel("Mean pixel intensity")
ax1.set_title("Mean correlation across all pairs per frame")
ax1.grid(True, alpha=0.3)

ax2.plot(snap_dates, frame_std, lw=0.8, color="darkorange", alpha=0.85)
ax2.set_ylabel("Std pixel intensity")
ax2.set_title("Std of correlation across all pairs per frame  "
              "(high = diverse market, low = all stocks moving together)")
ax2.grid(True, alpha=0.3)
ax2.xaxis.set_major_formatter(mdates.DateFormatter("%Y-%m"))
ax2.xaxis.set_major_locator(mdates.MonthLocator(interval=6))
plt.setp(ax2.xaxis.get_majorticklabels(), rotation=30, ha="right")

plt.tight_layout()
path3 = os.path.join(PROJECT_DIR, "plot_cell11_frame_stats.png")
plt.savefig(path3, dpi=150, bbox_inches="tight")
plt.show()
print(f"âœ“ Saved: {path3}")

print("\nâœ“ Cell 11 complete. Run Cell 12b for label construction.")
```

```python
# Cell 12b â€” Adaptive Resolution Dataset (DLSS-style zoom-in)
#
# Stable snapshots   â†’ daily resolution (stride=75 bars, from corr_video)
# Volatile snapshots â†’ sub-daily resolution (stride=18 bars, ~90min)
#                      built on-the-fly from pair_cache
#
# This is the "native high framerate" approach:
# - Normal market days: 1 frame per day is enough
# - Volatile market days: zoom in to 4-5 frames per day using real 5-min data
# - No synthetic data â€” real observations filling the gap
# - Model sees the actual trajectory through volatile periods
#   instead of just the start and end of a jump
#
# Each snapshot is represented as a dict with:
#   type      : "daily" or "subdaily"
#   t_idx     : bar index in the full 87122-bar series (for subdaily)
#   snap_idx  : index into corr_video (for daily)
#   frame_mse : volatility measure
#   is_val    : train or val

import numpy as np
import pickle, os, time
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from scipy.sparse import csr_matrix
from scipy.sparse.linalg import eigsh

PROJECT_DIR = r"C:\Users\HP\Documents\InterStockRelation_Paper\InterStockRelation"
CACHE_DIR   = os.path.join(PROJECT_DIR, "pair_cache")

# â”€â”€ Load video and metadata â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
print("Loading data...")
corr_video = np.load(os.path.join(PROJECT_DIR, "corr_video.npy"))

with open(os.path.join(PROJECT_DIR, "video_meta.pkl"), "rb") as f:
    vm = pickle.load(f)
with open(os.path.join(PROJECT_DIR, "pair_meta.pkl"), "rb") as f:
    pm = pickle.load(f)

snap_dates      = vm["snap_dates"]
snap_timesteps  = vm["snap_timesteps"]   # bar indices for daily snapshots
fiedler_order   = vm["fiedler_order"]
ordered_tickers = vm["ordered_tickers"]
N_STOCKS        = vm["n_stocks"]         # 95
N_SNAPS         = vm["n_snaps"]          # 1157

pair_meta  = pm["pairs"]
col_names  = pm["columns"]
COL        = {name: idx for idx, name in enumerate(col_names)}
timestamps = pm["timestamps"]
ticker_idx = {t: i for i, t in enumerate(vm["tickers"])}
pair_row   = np.array([ticker_idx[a] for a, b in pair_meta], dtype=np.int32)
pair_col   = np.array([ticker_idx[b] for a, b in pair_meta], dtype=np.int32)
R_EWMA_IDX = COL["R_ewma"]

print(f"âœ“ corr_video : {corr_video.shape}")
print(f"  N_STOCKS   : {N_STOCKS}")
print(f"  N_SNAPS    : {N_SNAPS}")

# â”€â”€ Config â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
HORIZON          = 5      # predict 5 snapshots ahead (in daily units)
LOOKBACK         = 20     # 20 frames of history
TRAIN_RATIO      = 0.8
STRIDE_DAILY     = 75     # bars per daily snapshot
STRIDE_SUBDAILY  = 18     # bars per subdaily snapshot (~90 min)
SUBDAILY_MULT    = STRIDE_DAILY // STRIDE_SUBDAILY   # ~4 subdaily per daily

n_valid = N_SNAPS - HORIZON
print(f"\nHorizon       : {HORIZON} daily snapshots")
print(f"Lookback      : {LOOKBACK} frames")
print(f"Subdaily mult : {SUBDAILY_MULT}x zoom on volatile days")

# â”€â”€ Compute per-snapshot frame MSE â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
print("\nComputing per-snapshot frame MSE...")
frame_mse = np.array([
    float(np.mean((corr_video[t] - corr_video[t + HORIZON]) ** 2))
    for t in range(n_valid)
])
volatility_threshold = float(np.median(frame_mse))
print(f"  Median MSE (threshold) : {volatility_threshold:.6f}")

# â”€â”€ Stratified split on daily snapshots â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
valid_snaps    = np.array([s for s in range(n_valid) if s >= LOOKBACK])
snap_mse       = frame_mse[valid_snaps]
volatile_snaps = valid_snaps[snap_mse >= volatility_threshold]
stable_snaps   = valid_snaps[snap_mse <  volatility_threshold]

def stratified_split(snaps, train_ratio):
    snaps_sorted = np.sort(snaps)
    n_train = int(len(snaps_sorted) * train_ratio)
    return snaps_sorted[:n_train].tolist(), snaps_sorted[n_train:].tolist()

vol_train, vol_val = stratified_split(volatile_snaps, TRAIN_RATIO)
sta_train, sta_val = stratified_split(stable_snaps,   TRAIN_RATIO)

daily_train = sorted(vol_train + sta_train)
daily_val   = sorted(vol_val   + sta_val)

print(f"\nâ”€â”€ Daily Snapshots â”€â”€")
print(f"  Train : {len(daily_train)}  "
      f"(volatile: {len(vol_train)}, stable: {len(sta_train)})")
print(f"  Val   : {len(daily_val)}  "
      f"(volatile: {len(vol_val)}, stable: {len(sta_val)})")

# â”€â”€ Load R_ewma into RAM for subdaily frame construction â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
print("\nLoading R_ewma into RAM for subdaily frames...")
t0 = time.time()
r_ewma_all = np.empty((len(pair_meta), len(timestamps)), dtype=np.float32)
for i in range(len(pair_meta)):
    arr = np.load(os.path.join(CACHE_DIR, f"pair_{i:05d}.npy"))
    r_ewma_all[i, :] = arr[:, R_EWMA_IDX]
print(f"âœ“ Loaded in {time.time()-t0:.0f}s")

def build_corr_frame(t_idx):
    """Build a single 95Ã—95 Fiedler-ordered correlation frame at bar t_idx."""
    r_vals = np.nan_to_num(r_ewma_all[:, t_idx], nan=0.0) / 100.0
    mat    = np.zeros((N_STOCKS, N_STOCKS), dtype=np.float32)
    mat[pair_row, pair_col] = r_vals
    mat[pair_col, pair_row] = r_vals
    np.fill_diagonal(mat, 1.0)
    return mat[np.ix_(fiedler_order, fiedler_order)]

# â”€â”€ Generate subdaily snapshots for volatile training days â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
# For each volatile daily snapshot in train:
#   - Generate SUBDAILY_MULT additional snapshots at stride=18 bars
#   - Each subdaily snapshot uses real R_ewma from pair_cache
#   - Lookback and horizon are expressed in subdaily units
#   - Future target = corr frame HORIZON*STRIDE_DAILY bars ahead
#
# We store subdaily snapshots as bar indices (t_idx) rather than snap indices
# The Dataset class will build frames on-the-fly from r_ewma_all

print("\nGenerating subdaily snapshot indices for volatile train days...")
t0 = time.time()

subdaily_train = []   # list of t_idx (bar indices) for subdaily train snapshots

for snap_i in vol_train:
    # Center bar index of this daily snapshot
    center_t = snap_timesteps[snap_i]

    # Generate subdaily indices around this daily snapshot
    # We spread SUBDAILY_MULT sub-snapshots within Â±STRIDE_DAILY/2 bars
    for k in range(SUBDAILY_MULT):
        # Offset within the daily window
        offset = (k - SUBDAILY_MULT // 2) * STRIDE_SUBDAILY
        t_sub  = int(center_t + offset)

        # Bounds check â€” need LOOKBACK*STRIDE_SUBDAILY history
        # and HORIZON*STRIDE_DAILY future
        min_t = LOOKBACK * STRIDE_SUBDAILY
        max_t = len(timestamps) - HORIZON * STRIDE_DAILY - 1
        if min_t <= t_sub <= max_t:
            subdaily_train.append(t_sub)

# Remove duplicates and sort
subdaily_train = sorted(set(subdaily_train))
print(f"âœ“ Generated {len(subdaily_train)} subdaily train snapshots "
      f"in {time.time()-t0:.1f}s")
print(f"  From {len(vol_train)} volatile daily snapshots "
      f"Ã— {SUBDAILY_MULT} = up to {len(vol_train)*SUBDAILY_MULT} subdaily")

# â”€â”€ Precompute subdaily frames â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
# We precompute the correlation frames for all subdaily t_idx values
# and store them in a dict {t_idx: (95,95) frame}
# This avoids recomputing on every dataset access during training

print("\nPrecomputing subdaily correlation frames...")
print(f"  Need frames for {len(subdaily_train)} snapshots Ã— "
      f"{LOOKBACK} lookback + 1 future = up to "
      f"{len(subdaily_train) * (LOOKBACK + 1):,} frame evaluations")
print("  Using vectorized computation...")

# Collect all unique bar indices we need
t0 = time.time()
all_t_needed = set()
for t_sub in subdaily_train:
    # Lookback frames at subdaily stride
    for k in range(LOOKBACK):
        all_t_needed.add(t_sub - (LOOKBACK - k) * STRIDE_SUBDAILY)
    # Future target frame
    all_t_needed.add(t_sub + HORIZON * STRIDE_DAILY)

all_t_needed = sorted(all_t_needed)
print(f"  Unique bar indices needed : {len(all_t_needed):,}")

# Build all needed frames
subdaily_frame_cache = {}
for t_idx in all_t_needed:
    if 0 <= t_idx < len(timestamps):
        subdaily_frame_cache[t_idx] = build_corr_frame(t_idx)

print(f"âœ“ Built {len(subdaily_frame_cache):,} frames in {time.time()-t0:.1f}s")

# â”€â”€ Compute baselines â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
print("\nComputing baselines on stratified val set...")
persistence_mse = float(np.mean([
    np.mean((corr_video[t] - corr_video[t + HORIZON]) ** 2)
    for t in daily_val
]))
train_mean_frame = corr_video[daily_train].mean(axis=0)
mean_frame_mse   = float(np.mean([
    np.mean((train_mean_frame - corr_video[t + HORIZON]) ** 2)
    for t in daily_val
]))
print(f"  Persistence MSE : {persistence_mse:.6f}")
print(f"  Mean-frame MSE  : {mean_frame_mse:.6f}")

# â”€â”€ Save everything â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
print("\nSaving dataset metadata...")
dataset_meta = {
    # Config
    "horizon"             : HORIZON,
    "lookback"            : LOOKBACK,
    "train_ratio"         : TRAIN_RATIO,
    "stride_daily"        : STRIDE_DAILY,
    "stride_subdaily"     : STRIDE_SUBDAILY,
    # Snapshot lists
    "daily_train"         : daily_train,
    "daily_val"           : daily_val,
    "subdaily_train"      : subdaily_train,
    # Frame cache
    "subdaily_frame_cache": subdaily_frame_cache,
    # Stats
    "n_stocks"            : N_STOCKS,
    "n_snaps"             : N_SNAPS,
    "snap_dates"          : snap_dates,
    "snap_timesteps"      : snap_timesteps,
    "ordered_tickers"     : ordered_tickers,
    "fiedler_order"       : fiedler_order,
    "frame_mse"           : frame_mse,
    "volatility_threshold": volatility_threshold,
    "persistence_mse"     : persistence_mse,
    "mean_frame_mse"      : mean_frame_mse,
    "train_mean_frame"    : train_mean_frame,
}
with open(os.path.join(PROJECT_DIR, "dataset_meta_v2.pkl"), "wb") as f:
    pickle.dump(dataset_meta, f)

print(f"âœ“ Saved dataset_meta_v2.pkl")

# â”€â”€ Summary â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
total_train = len(daily_train) + len(subdaily_train)
print(f"\nâ”€â”€ Dataset Summary â”€â”€")
print(f"  Daily train snaps    : {len(daily_train)}")
print(f"  Subdaily train snaps : {len(subdaily_train)}  "
      f"(zoom-in on {len(vol_train)} volatile days)")
print(f"  Total train snaps    : {total_train}  "
      f"({len(subdaily_train)/total_train*100:.1f}% subdaily)")
print(f"  Val snaps            : {len(daily_val)}  (daily only)")
print(f"  Persistence MSE      : {persistence_mse:.6f}")
print(f"  Mean-frame MSE       : {mean_frame_mse:.6f}")

# â”€â”€ Plot: daily vs subdaily snapshot density over time â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(14, 7), sharex=True)
fig.suptitle("Adaptive Resolution Dataset â€” DLSS-style Zoom-in on Volatile Days",
             fontsize=12, fontweight="bold")

all_dates = [snap_dates[t] for t in range(n_valid)]
ax1.plot(all_dates, frame_mse, lw=0.8, color="steelblue", alpha=0.8)
ax1.axhline(volatility_threshold, color="red", lw=1.5, linestyle="--",
            label=f"Volatility threshold = {volatility_threshold:.4f}")
ax1.set_ylabel("Frame MSE"); ax1.legend(fontsize=9); ax1.grid(True, alpha=0.3)
ax1.set_title("Market volatility over time")

# Show subdaily snapshot density
subdaily_dates = [timestamps[t] for t in subdaily_train]
daily_dates    = [snap_dates[s] for s in daily_train]

ax2.scatter(daily_dates,    [0.3] * len(daily_dates),
            s=2, color="steelblue",  alpha=0.4, label=f"Daily ({len(daily_train)})")
ax2.scatter(subdaily_dates, [0.7] * len(subdaily_dates),
            s=1, color="crimson",    alpha=0.3, label=f"Subdaily ({len(subdaily_train)})")
ax2.set_ylabel("Snapshot type"); ax2.set_ylim(0, 1)
ax2.set_yticks([0.3, 0.7])
ax2.set_yticklabels(["Daily", "Subdaily"])
ax2.legend(fontsize=9); ax2.grid(True, alpha=0.3)
ax2.set_title("Training snapshot distribution â€” subdaily zoom on volatile periods")
ax2.xaxis.set_major_formatter(mdates.DateFormatter("%Y-%m"))
ax2.xaxis.set_major_locator(mdates.MonthLocator(interval=6))
plt.setp(ax2.xaxis.get_majorticklabels(), rotation=30, ha="right")

plt.tight_layout()
path = os.path.join(PROJECT_DIR, "plot_cell12b_adaptive_resolution.png")
plt.savefig(path, dpi=150, bbox_inches="tight")
plt.show()
print(f"âœ“ Saved: {path}")
print("\nâœ“ Cell 12b complete. Run Cell 13 for CNN+LSTM with adaptive resolution.")
```

```python
# Cell 13 â€” CNN+LSTM Regression with Adaptive Resolution Training
#
# Same architecture as before â€” CNN per frame, LSTM across frames.
# Key change: Dataset now handles both daily and subdaily snapshots.
#   - Daily snapshots  â†’ frames from corr_video (precomputed)
#   - Subdaily snapshots â†’ frames from subdaily_frame_cache (precomputed in 12b)
#
# The model sees 3x more training signal on volatile periods
# without any synthetic data â€” real 90-min resolution frames.

import numpy as np
import pickle, os, time
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from tqdm import tqdm
from scipy.stats import pearsonr
import matplotlib.pyplot as plt

PROJECT_DIR = r"C:\Users\HP\Documents\InterStockRelation_Paper\InterStockRelation"
MODELS_DIR  = os.path.join(PROJECT_DIR, "models")
os.makedirs(MODELS_DIR, exist_ok=True)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Device : {device}  ({torch.cuda.get_device_name(0)})")

# â”€â”€ Load data â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
print("\nLoading data...")
corr_video = np.load(os.path.join(PROJECT_DIR, "corr_video.npy"))

with open(os.path.join(PROJECT_DIR, "dataset_meta_v2.pkl"), "rb") as f:
    dm = pickle.load(f)

N_STOCKS           = dm["n_stocks"]
N_SNAPS            = dm["n_snaps"]
HORIZON            = dm["horizon"]
LOOKBACK           = dm["lookback"]
STRIDE_DAILY       = dm["stride_daily"]
STRIDE_SUBDAILY    = dm["stride_subdaily"]
daily_train        = dm["daily_train"]
daily_val          = dm["daily_val"]
subdaily_train     = dm["subdaily_train"]
subdaily_frame_cache= dm["subdaily_frame_cache"]
frame_mse          = dm["frame_mse"]
persistence_mse    = dm["persistence_mse"]
mean_frame_mse     = dm["mean_frame_mse"]
train_mean_frame   = dm["train_mean_frame"]
snap_dates         = dm["snap_dates"]
snap_timesteps     = dm["snap_timesteps"]
volatility_threshold= dm["volatility_threshold"]

print(f"âœ“ corr_video       : {corr_video.shape}")
print(f"  Daily train      : {len(daily_train)}")
print(f"  Subdaily train   : {len(subdaily_train)}")
print(f"  Val snaps        : {len(daily_val)}")
print(f"  Persistence MSE  : {persistence_mse:.6f}")
print(f"  Mean-frame MSE   : {mean_frame_mse:.6f}")

# â”€â”€ Dataset â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
class AdaptiveResDataset(Dataset):
    """
    Handles both daily and subdaily snapshots.

    Daily snapshot (snap_idx):
      - Lookback frames from corr_video[snap_idx-LOOKBACK : snap_idx]
      - Future target  = corr_video[snap_idx + HORIZON]
      - Volatility weight = frame_mse[snap_idx]

    Subdaily snapshot (t_idx â€” bar index):
      - Lookback frames from subdaily_frame_cache at subdaily stride
      - Future target = subdaily_frame_cache[t_idx + HORIZON*STRIDE_DAILY]
      - Volatility weight = MSE between current and future frame
    """
    def __init__(self, daily_snaps, subdaily_snaps=None):
        self.samples = []

        # Daily samples: (type, index)
        for s in daily_snaps:
            self.samples.append(("daily", s))

        # Subdaily samples
        if subdaily_snaps:
            for t in subdaily_snaps:
                self.samples.append(("subdaily", t))

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        kind, index = self.samples[idx]

        if kind == "daily":
            # Frames from corr_video
            frames  = corr_video[index - LOOKBACK: index, :, :]  # (L, 95, 95)
            future  = corr_video[index + HORIZON]                 # (95, 95)
            weight  = float(frame_mse[index])

        else:  # subdaily
            # Frames from precomputed cache at subdaily stride
            t_idx   = index
            frame_list = []
            for k in range(LOOKBACK):
                t_k = t_idx - (LOOKBACK - k) * STRIDE_SUBDAILY
                if t_k in subdaily_frame_cache:
                    frame_list.append(subdaily_frame_cache[t_k])
                else:
                    # Fallback: zeros if bar not in cache
                    frame_list.append(np.zeros((N_STOCKS, N_STOCKS),
                                               dtype=np.float32))
            frames = np.stack(frame_list, axis=0)               # (L, 95, 95)

            # Future target
            t_future = t_idx + HORIZON * STRIDE_DAILY
            if t_future in subdaily_frame_cache:
                future = subdaily_frame_cache[t_future]
            else:
                future = np.zeros((N_STOCKS, N_STOCKS), dtype=np.float32)

            weight = float(np.mean((frames[-1] - future) ** 2))

        return (
            torch.from_numpy(frames.copy()).float(),
            torch.from_numpy(future.copy()).float(),
            torch.tensor(weight, dtype=torch.float32),
        )

BATCH_SIZE = 16
train_ds   = AdaptiveResDataset(daily_train, subdaily_train)
val_ds     = AdaptiveResDataset(daily_val,   None)
train_dl   = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True,
                        num_workers=0, pin_memory=True)
val_dl     = DataLoader(val_ds,   batch_size=BATCH_SIZE, shuffle=False,
                        num_workers=0, pin_memory=True)

print(f"\nTrain samples : {len(train_ds)}  "
      f"({len(daily_train)} daily + {len(subdaily_train)} subdaily)")
print(f"Val samples   : {len(val_ds)}  (daily only)")

# â”€â”€ Architecture (unchanged from previous Cell 13) â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
class FrameCNN(nn.Module):
    def __init__(self, cnn_out_dim=256):
        super().__init__()
        self.conv1 = nn.Conv2d(1,  16, kernel_size=5, padding=2)
        self.conv2 = nn.Conv2d(16, 32, kernel_size=5, padding=2)
        self.conv3 = nn.Conv2d(32, 64, kernel_size=3, padding=1)
        self.bn1   = nn.BatchNorm2d(16)
        self.bn2   = nn.BatchNorm2d(32)
        self.bn3   = nn.BatchNorm2d(64)
        self.pool  = nn.MaxPool2d(2, 2)
        self.drop  = nn.Dropout2d(0.2)
        self.apool = nn.AdaptiveAvgPool2d((6, 6))
        self.fc    = nn.Linear(64 * 6 * 6, cnn_out_dim)
        self.norm  = nn.LayerNorm(cnn_out_dim)

    def forward(self, x):
        x = F.relu(self.bn1(self.conv1(x))); x = self.pool(x)
        x = F.relu(self.bn2(self.conv2(x))); x = self.pool(x)
        x = self.drop(x)
        x = F.relu(self.bn3(self.conv3(x)))
        x = self.apool(x).flatten(1)
        return self.norm(F.relu(self.fc(x)))


class CorrVideoReg(nn.Module):
    def __init__(self, cnn_out_dim=256, lstm_hidden=256,
                 lstm_layers=2, n_stocks=95, dropout=0.3):
        super().__init__()
        self.n_stocks = n_stocks
        self.cnn      = FrameCNN(cnn_out_dim)
        self.lstm     = nn.LSTM(cnn_out_dim, lstm_hidden, lstm_layers,
                                batch_first=True,
                                dropout=dropout if lstm_layers > 1 else 0.0)
        self.norm     = nn.LayerNorm(lstm_hidden)
        self.dropout  = nn.Dropout(dropout)
        self.decoder  = nn.Sequential(
            nn.Linear(lstm_hidden, 512),  nn.ReLU(), nn.Dropout(dropout),
            nn.Linear(512, 1024),         nn.ReLU(), nn.Dropout(dropout),
            nn.Linear(1024, n_stocks * n_stocks), nn.Tanh()
        )

    def forward(self, frames):
        B, L, H, W  = frames.shape
        cnn_out     = self.cnn(frames.reshape(B*L, 1, H, W))
        cnn_seq     = cnn_out.reshape(B, L, -1)
        lstm_out, _ = self.lstm(cnn_seq)
        h           = self.dropout(self.norm(lstm_out[:, -1, :]))
        return self.decoder(h).reshape(B, self.n_stocks, self.n_stocks)


def weighted_mse(pred, target, weights):
    per_sample = ((pred - target) ** 2).mean(dim=(1, 2))
    w = weights / weights.sum() * len(weights)
    return (w * per_sample).mean()

# â”€â”€ Training â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
MAX_EPOCHS = 40
LR         = 1e-3
PATIENCE   = 8

model     = CorrVideoReg(n_stocks=N_STOCKS).to(device)
optimizer = torch.optim.Adam(model.parameters(), lr=LR, weight_decay=1e-4)
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer, patience=4, factor=0.5, verbose=False)

print(f"\nParameters : {sum(p.numel() for p in model.parameters()):,}")
print(f"Baselines  : persistence={persistence_mse:.4f}  "
      f"mean-frame={mean_frame_mse:.4f}\n")

best_val_mse = float("inf")
patience_ctr = 0
best_path    = os.path.join(MODELS_DIR, "cnn_lstm_reg_best.pt")
history      = []

for epoch in range(1, MAX_EPOCHS + 1):
    t0 = time.time()

    # Train
    model.train()
    tr_loss = tr_total = 0.0
    for frames, future, w in tqdm(train_dl,
            desc=f"Ep {epoch:>2} Train", ncols=80, leave=False):
        frames, future, w = frames.to(device), future.to(device), w.to(device)
        optimizer.zero_grad()
        pred = model(frames)
        loss = weighted_mse(pred, future, w)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        tr_loss  += loss.item() * frames.size(0)
        tr_total += frames.size(0)

    # Validate
    model.eval()
    va_loss = va_total = 0.0
    all_preds, all_futures = [], []
    with torch.no_grad():
        for frames, future, w in tqdm(val_dl,
                desc=f"Ep {epoch:>2} Val  ", ncols=80, leave=False):
            frames, future, w = frames.to(device), future.to(device), w.to(device)
            pred     = model(frames)
            loss     = weighted_mse(pred, future, w)
            va_loss += loss.item() * frames.size(0)
            va_total+= frames.size(0)
            all_preds.append(pred.cpu().numpy())
            all_futures.append(future.cpu().numpy())

    tr_mse = tr_loss / tr_total
    va_mse = va_loss / va_total
    lr_now = optimizer.param_groups[0]["lr"]

    pf = np.concatenate(all_preds).reshape(-1)
    ff = np.concatenate(all_futures).reshape(-1)
    pearson_r, _ = pearsonr(pf[:100000], ff[:100000])

    vs_persist = (persistence_mse - va_mse) / persistence_mse * 100
    vs_mean    = (mean_frame_mse  - va_mse) / mean_frame_mse  * 100
    improved   = " âœ“" if va_mse < best_val_mse else ""

    scheduler.step(va_mse)
    history.append(dict(epoch=epoch, tr_mse=tr_mse, va_mse=va_mse,
                        pearson_r=pearson_r))

    print(f"Ep {epoch:>2}/{MAX_EPOCHS} | "
          f"tr={tr_mse:.5f} | va={va_mse:.5f} | "
          f"r={pearson_r:.4f} | "
          f"vs_persist={vs_persist:+.1f}% | "
          f"vs_mean={vs_mean:+.1f}% | "
          f"lr={lr_now:.1e} | {time.time()-t0:.0f}s{improved}")

    if va_mse < best_val_mse:
        best_val_mse = va_mse
        patience_ctr = 0
        torch.save(model.state_dict(), best_path)
    else:
        patience_ctr += 1
        if patience_ctr >= PATIENCE:
            print(f"\nEarly stopping at epoch {epoch}")
            break

print(f"\nâœ“ Best val MSE : {best_val_mse:.6f}")
print(f"  vs persistence : {(persistence_mse-best_val_mse)/persistence_mse*100:+.1f}%")
print(f"  vs mean-frame  : {(mean_frame_mse-best_val_mse)/mean_frame_mse*100:+.1f}%")

# â”€â”€ Final evaluation â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
model.load_state_dict(torch.load(best_path, map_location=device))
model.eval()
all_preds, all_futures, all_weights = [], [], []
with torch.no_grad():
    for frames, future, w in val_dl:
        pred = model(frames.to(device)).cpu().numpy()
        all_preds.append(pred)
        all_futures.append(future.numpy())
        all_weights.extend(w.numpy())

all_preds   = np.concatenate(all_preds,   axis=0)
all_futures = np.concatenate(all_futures, axis=0)
all_weights = np.array(all_weights)

per_snap_mse = ((all_preds - all_futures) ** 2).mean(axis=(1, 2))

# Directional accuracy
val_current = np.array([corr_video[s] for s in daily_val[:len(all_preds)]])
dir_acc     = (np.sign(all_preds - val_current) ==
               np.sign(all_futures - val_current)).mean()

pf = all_preds.reshape(-1);  ff = all_futures.reshape(-1)
pearson_r_final, _ = pearsonr(pf[:500000], ff[:500000])

volatile_mask = all_weights >= volatility_threshold
mse_volatile  = per_snap_mse[volatile_mask].mean()
mse_stable    = per_snap_mse[~volatile_mask].mean()

print(f"\nâ”€â”€ Final Evaluation â”€â”€")
print(f"  Val MSE (unweighted)  : {per_snap_mse.mean():.6f}  "
      f"(prev: 0.073406)")
print(f"  Persistence MSE       : {persistence_mse:.6f}")
print(f"  Mean-frame MSE        : {mean_frame_mse:.6f}")
print(f"  Pearson r             : {pearson_r_final:.4f}  (prev: 0.2769)")
print(f"  Directional accuracy  : {dir_acc*100:.2f}%  (prev: 72.88%)")
print(f"  MSE on volatile snaps : {mse_volatile:.6f}  (prev: 0.111714)")
print(f"  MSE on stable snaps   : {mse_stable:.6f}  (prev: 0.036419)")

# â”€â”€ Save â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
np.save(os.path.join(PROJECT_DIR, "cnn_lstm_val_preds.npy"), all_preds)
with open(os.path.join(MODELS_DIR, "cnn_lstm_reg_history.pkl"), "wb") as f:
    pickle.dump(history, f)
print(f"\nâœ“ Saved: {best_path}")

# â”€â”€ Release GPU â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
del model
torch.cuda.empty_cache()
print(f"âœ“ GPU memory released")

# â”€â”€ Plot â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
fig.suptitle("CNN+LSTM Regression â€” Adaptive Resolution Training",
             fontsize=13, fontweight="bold")

epochs  = [h["epoch"]     for h in history]
tr_mses = [h["tr_mse"]    for h in history]
va_mses = [h["va_mse"]    for h in history]
rs      = [h["pearson_r"] for h in history]

axes[0].plot(epochs, tr_mses, "o-", color="steelblue",  label="Train MSE")
axes[0].plot(epochs, va_mses, "o-", color="darkorange", label="Val MSE")
axes[0].axhline(persistence_mse, color="red",   lw=1.5,
                linestyle="--", label=f"Persistence {persistence_mse:.4f}")
axes[0].axhline(mean_frame_mse,  color="green", lw=1.5,
                linestyle="--", label=f"Mean-frame {mean_frame_mse:.4f}")
axes[0].set_title("MSE over epochs")
axes[0].set_xlabel("Epoch"); axes[0].set_ylabel("MSE")
axes[0].legend(fontsize=8); axes[0].grid(True, alpha=0.3)

axes[1].plot(epochs, rs, "o-", color="forestgreen")
axes[1].set_title("Pearson r (predicted vs actual)")
axes[1].set_xlabel("Epoch"); axes[1].set_ylabel("Pearson r")
axes[1].set_ylim(0, 1); axes[1].grid(True, alpha=0.3)

sample_idx = np.random.choice(len(pf[:500000]), 5000, replace=False)
axes[2].scatter(ff[:500000][sample_idx], pf[:500000][sample_idx],
                alpha=0.1, s=1, color="steelblue")
axes[2].plot([-1,1], [-1,1], "r--", lw=1.5, label="Perfect")
axes[2].set_xlabel("Actual R_ewma"); axes[2].set_ylabel("Predicted R_ewma")
axes[2].set_title(f"Predicted vs Actual (r={pearson_r_final:.3f})")
axes[2].legend(fontsize=9); axes[2].grid(True, alpha=0.3)

plt.tight_layout()
path = os.path.join(PROJECT_DIR, "plot_cell13_cnn_lstm_adaptive.png")
plt.savefig(path, dpi=150, bbox_inches="tight")
plt.show()
print(f"âœ“ Saved: {path}")
print("\nâœ“ Cell 13 complete. Run Cell 14 for 3D CNN.")
```

```python
# Cell 14 â€” 3D CNN Pure Regression with Adaptive Resolution
#
# 3D conv kernels slide across space AND time simultaneously.
# A (T, H, W) kernel detects spatial patterns changing over T frames â€”
# directly on the raw pixel values, not on compressed feature vectors.
#
# Key advantage over CNN+LSTM:
#   Spatial structure is preserved through time.
#   The kernel sees "cluster block at rows 10-15 brightening over 3 days"
#   as a single operation â€” no information lost to sequential compression.
#
# Same adaptive resolution dataset as Cell 13.
# Same evaluation metrics for direct comparison.

import numpy as np
import pickle, os, time
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from tqdm import tqdm
from scipy.stats import pearsonr
import matplotlib.pyplot as plt

PROJECT_DIR = r"C:\Users\HP\Documents\InterStockRelation_Paper\InterStockRelation"
MODELS_DIR  = os.path.join(PROJECT_DIR, "models")

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Device : {device}  ({torch.cuda.get_device_name(0)})")

# â”€â”€ Load data â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
print("\nLoading data...")
corr_video = np.load(os.path.join(PROJECT_DIR, "corr_video.npy"))

with open(os.path.join(PROJECT_DIR, "dataset_meta_v2.pkl"), "rb") as f:
    dm = pickle.load(f)

N_STOCKS            = dm["n_stocks"]
HORIZON             = dm["horizon"]
LOOKBACK            = dm["lookback"]
STRIDE_DAILY        = dm["stride_daily"]
STRIDE_SUBDAILY     = dm["stride_subdaily"]
daily_train         = dm["daily_train"]
daily_val           = dm["daily_val"]
subdaily_train      = dm["subdaily_train"]
subdaily_frame_cache= dm["subdaily_frame_cache"]
frame_mse           = dm["frame_mse"]
persistence_mse     = dm["persistence_mse"]
mean_frame_mse      = dm["mean_frame_mse"]
volatility_threshold= dm["volatility_threshold"]
snap_dates          = dm["snap_dates"]

print(f"âœ“ corr_video      : {corr_video.shape}")
print(f"  Daily train     : {len(daily_train)}")
print(f"  Subdaily train  : {len(subdaily_train)}")
print(f"  Val snaps       : {len(daily_val)}")
print(f"  Persistence MSE : {persistence_mse:.6f}")
print(f"  Mean-frame MSE  : {mean_frame_mse:.6f}")

# â”€â”€ Dataset (same as Cell 13) â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
class AdaptiveResDataset(Dataset):
    def __init__(self, daily_snaps, subdaily_snaps=None):
        self.samples = [("daily", s) for s in daily_snaps]
        if subdaily_snaps:
            self.samples += [("subdaily", t) for t in subdaily_snaps]

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        kind, index = self.samples[idx]
        if kind == "daily":
            frames = corr_video[index - LOOKBACK: index, :, :]
            future = corr_video[index + HORIZON]
            weight = float(frame_mse[index])
        else:
            t_idx  = index
            frames = np.stack([
                subdaily_frame_cache.get(
                    t_idx - (LOOKBACK - k) * STRIDE_SUBDAILY,
                    np.zeros((N_STOCKS, N_STOCKS), dtype=np.float32))
                for k in range(LOOKBACK)
            ], axis=0)
            t_future = t_idx + HORIZON * STRIDE_DAILY
            future   = subdaily_frame_cache.get(
                t_future, np.zeros((N_STOCKS, N_STOCKS), dtype=np.float32))
            weight   = float(np.mean((frames[-1] - future) ** 2))
        return (
            torch.from_numpy(frames.copy()).float(),
            torch.from_numpy(np.array(future).copy()).float(),
            torch.tensor(weight, dtype=torch.float32),
        )

# Smaller batch size for 3D conv â€” more memory intensive
BATCH_SIZE = 8
train_ds   = AdaptiveResDataset(daily_train, subdaily_train)
val_ds     = AdaptiveResDataset(daily_val,   None)
train_dl   = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True,
                        num_workers=0, pin_memory=True)
val_dl     = DataLoader(val_ds,   batch_size=BATCH_SIZE, shuffle=False,
                        num_workers=0, pin_memory=True)

print(f"\nTrain samples : {len(train_ds)}")
print(f"Val samples   : {len(val_ds)}")
print(f"Batch size    : {BATCH_SIZE}  (smaller due to 3D conv memory)")

# â”€â”€ Architecture â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€

class Corr3DCNN(nn.Module):
    """
    Pure 3D CNN for correlation matrix video regression.

    Input : (B, 1, LOOKBACK, 95, 95)
    Output: (B, 95, 95) â€” predicted future correlation matrix

    3D kernels detect spatiotemporal patterns directly:
      conv1 (3,3,3) â€” short-range: local cluster changing over 3 days
      conv2 (3,5,5) â€” medium spatial, short temporal: broader structure
      conv3 (5,3,3) â€” longer temporal: slower-moving patterns
      conv4 (3,3,3) â€” deep features

    No LSTM needed â€” temporal context baked into 3D kernels.
    """
    def __init__(self, n_stocks=95, dropout=0.3):
        super().__init__()
        self.n_stocks = n_stocks

        # Block 1: short-range spatiotemporal
        self.conv1 = nn.Conv3d(1,  16, kernel_size=(3,3,3), padding=(1,1,1))
        self.bn1   = nn.BatchNorm3d(16)

        # Block 2: broader spatial context
        self.conv2 = nn.Conv3d(16, 32, kernel_size=(3,5,5), padding=(1,2,2))
        self.bn2   = nn.BatchNorm3d(32)

        # Block 3: longer temporal patterns
        self.conv3 = nn.Conv3d(32, 64, kernel_size=(5,3,3), padding=(2,1,1))
        self.bn3   = nn.BatchNorm3d(64)

        # Block 4: deep feature extraction
        self.conv4 = nn.Conv3d(64, 128, kernel_size=(3,3,3), padding=(1,1,1))
        self.bn4   = nn.BatchNorm3d(128)

        # Collapse temporal dim, pool spatial
        # AdaptiveAvgPool3d â†’ (B, 128, 1, 6, 6) â†’ flatten â†’ 4608
        self.temporal_pool = nn.AdaptiveAvgPool3d((1, 6, 6))
        self.dropout       = nn.Dropout(dropout)
        self.fc            = nn.Linear(128 * 6 * 6, 512)
        self.fc_norm       = nn.LayerNorm(512)

        # Decoder: predict future 95Ã—95 correlation matrix
        self.decoder = nn.Sequential(
            nn.Linear(512, 512),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(512, 1024),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(1024, n_stocks * n_stocks),
            nn.Tanh()   # constrain to [-1, +1]
        )

    def forward(self, frames):
        # frames: (B, LOOKBACK, 95, 95)
        B = frames.size(0)
        x = frames.unsqueeze(1)                    # (B, 1, L, 95, 95)

        x = F.relu(self.bn1(self.conv1(x)))        # (B, 16,  L, 95, 95)
        x = F.relu(self.bn2(self.conv2(x)))        # (B, 32,  L, 95, 95)
        x = F.relu(self.bn3(self.conv3(x)))        # (B, 64,  L, 95, 95)
        x = F.relu(self.bn4(self.conv4(x)))        # (B, 128, L, 95, 95)

        x = self.temporal_pool(x)                  # (B, 128, 1, 6, 6)
        x = self.dropout(x.flatten(1))             # (B, 4608)
        x = self.fc_norm(F.relu(self.fc(x)))       # (B, 512)
        x = self.dropout(x)

        out = self.decoder(x)                      # (B, 95*95)
        return out.reshape(B, self.n_stocks, self.n_stocks)


def weighted_mse(pred, target, weights):
    per_sample = ((pred - target) ** 2).mean(dim=(1, 2))
    w = weights / weights.sum() * len(weights)
    return (w * per_sample).mean()

# â”€â”€ Training â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
MAX_EPOCHS = 40
LR         = 1e-3
PATIENCE   = 8

torch.cuda.empty_cache()
model     = Corr3DCNN(n_stocks=N_STOCKS).to(device)
optimizer = torch.optim.Adam(model.parameters(), lr=LR, weight_decay=1e-4)
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer, patience=4, factor=0.5, verbose=False)

print(f"\nParameters : {sum(p.numel() for p in model.parameters()):,}")
print(f"Baselines  : persistence={persistence_mse:.4f}  "
      f"mean-frame={mean_frame_mse:.4f}\n")

best_val_mse = float("inf")
patience_ctr = 0
best_path    = os.path.join(MODELS_DIR, "3dcnn_reg_best.pt")
history      = []

for epoch in range(1, MAX_EPOCHS + 1):
    t0 = time.time()

    # Train
    model.train()
    tr_loss = tr_total = 0.0
    for frames, future, w in tqdm(train_dl,
            desc=f"Ep {epoch:>2} Train", ncols=80, leave=False):
        frames, future, w = frames.to(device), future.to(device), w.to(device)
        optimizer.zero_grad()
        pred = model(frames)
        loss = weighted_mse(pred, future, w)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        tr_loss  += loss.item() * frames.size(0)
        tr_total += frames.size(0)

    # Validate
    model.eval()
    va_loss = va_total = 0.0
    all_preds, all_futures, all_weights = [], [], []
    with torch.no_grad():
        for frames, future, w in tqdm(val_dl,
                desc=f"Ep {epoch:>2} Val  ", ncols=80, leave=False):
            frames, future, w = frames.to(device), future.to(device), w.to(device)
            pred     = model(frames)
            loss     = weighted_mse(pred, future, w)
            va_loss += loss.item() * frames.size(0)
            va_total+= frames.size(0)
            all_preds.append(pred.cpu().numpy())
            all_futures.append(future.cpu().numpy())
            all_weights.extend(w.cpu().numpy())

    tr_mse = tr_loss / tr_total
    va_mse = va_loss / va_total
    lr_now = optimizer.param_groups[0]["lr"]

    pf = np.concatenate(all_preds).reshape(-1)
    ff = np.concatenate(all_futures).reshape(-1)
    pearson_r, _ = pearsonr(pf[:100000], ff[:100000])

    vs_persist = (persistence_mse - va_mse) / persistence_mse * 100
    vs_mean    = (mean_frame_mse  - va_mse) / mean_frame_mse  * 100
    improved   = " âœ“" if va_mse < best_val_mse else ""

    scheduler.step(va_mse)
    history.append(dict(epoch=epoch, tr_mse=tr_mse, va_mse=va_mse,
                        pearson_r=pearson_r))

    print(f"Ep {epoch:>2}/{MAX_EPOCHS} | "
          f"tr={tr_mse:.5f} | va={va_mse:.5f} | "
          f"r={pearson_r:.4f} | "
          f"vs_persist={vs_persist:+.1f}% | "
          f"vs_mean={vs_mean:+.1f}% | "
          f"lr={lr_now:.1e} | {time.time()-t0:.0f}s{improved}")

    if va_mse < best_val_mse:
        best_val_mse = va_mse
        patience_ctr = 0
        torch.save(model.state_dict(), best_path)
    else:
        patience_ctr += 1
        if patience_ctr >= PATIENCE:
            print(f"\nEarly stopping at epoch {epoch}")
            break

print(f"\nâœ“ Best val MSE : {best_val_mse:.6f}")
print(f"  vs persistence : {(persistence_mse-best_val_mse)/persistence_mse*100:+.1f}%")
print(f"  vs mean-frame  : {(mean_frame_mse-best_val_mse)/mean_frame_mse*100:+.1f}%")

# â”€â”€ Final evaluation â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
model.load_state_dict(torch.load(best_path, map_location=device))
model.eval()
all_preds, all_futures, all_weights = [], [], []
with torch.no_grad():
    for frames, future, w in val_dl:
        pred = model(frames.to(device)).cpu().numpy()
        all_preds.append(pred)
        all_futures.append(future.numpy())
        all_weights.extend(w.numpy())

all_preds    = np.concatenate(all_preds,   axis=0)
all_futures  = np.concatenate(all_futures, axis=0)
all_weights  = np.array(all_weights)
per_snap_mse = ((all_preds - all_futures) ** 2).mean(axis=(1, 2))

val_current  = np.array([corr_video[s] for s in daily_val[:len(all_preds)]])
dir_acc      = (np.sign(all_preds - val_current) ==
                np.sign(all_futures - val_current)).mean()

pf = all_preds.reshape(-1);  ff = all_futures.reshape(-1)
pearson_r_final, _ = pearsonr(pf[:500000], ff[:500000])

volatile_mask = all_weights >= volatility_threshold
mse_volatile  = per_snap_mse[volatile_mask].mean() if volatile_mask.any() else float("nan")
mse_stable    = per_snap_mse[~volatile_mask].mean() if (~volatile_mask).any() else float("nan")

print(f"\nâ”€â”€ Final Evaluation â”€â”€")
print(f"  Val MSE (unweighted)  : {per_snap_mse.mean():.6f}  (CNN+LSTM: 0.072454)")
print(f"  Persistence MSE       : {persistence_mse:.6f}")
print(f"  Mean-frame MSE        : {mean_frame_mse:.6f}")
print(f"  Pearson r             : {pearson_r_final:.4f}  (CNN+LSTM: 0.2887)")
print(f"  Directional accuracy  : {dir_acc*100:.2f}%  (CNN+LSTM: 73.03%)")
print(f"  MSE on volatile snaps : {mse_volatile:.6f}  (CNN+LSTM: 0.109566)")
print(f"  MSE on stable snaps   : {mse_stable:.6f}  (CNN+LSTM: 0.036622)")

# â”€â”€ Save â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
np.save(os.path.join(PROJECT_DIR, "3dcnn_val_preds.npy"), all_preds)
with open(os.path.join(MODELS_DIR, "3dcnn_reg_history.pkl"), "wb") as f:
    pickle.dump(history, f)
print(f"\nâœ“ Saved: {best_path}")

# â”€â”€ Release GPU â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
del model
torch.cuda.empty_cache()
print(f"âœ“ GPU memory released")

# â”€â”€ Plot â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
fig.suptitle("3D CNN Regression â€” Adaptive Resolution Training",
             fontsize=13, fontweight="bold")

epochs  = [h["epoch"]     for h in history]
tr_mses = [h["tr_mse"]    for h in history]
va_mses = [h["va_mse"]    for h in history]
rs      = [h["pearson_r"] for h in history]

axes[0].plot(epochs, tr_mses, "o-", color="steelblue",  label="Train MSE")
axes[0].plot(epochs, va_mses, "o-", color="darkorange", label="Val MSE")
axes[0].axhline(persistence_mse, color="red",   lw=1.5,
                linestyle="--", label=f"Persistence {persistence_mse:.4f}")
axes[0].axhline(mean_frame_mse,  color="green", lw=1.5,
                linestyle="--", label=f"Mean-frame {mean_frame_mse:.4f}")
# Reference CNN+LSTM best
axes[0].axhline(0.075715, color="purple", lw=1.2,
                linestyle=":", label="CNN+LSTM best 0.0757")
axes[0].set_title("MSE over epochs")
axes[0].set_xlabel("Epoch"); axes[0].set_ylabel("MSE")
axes[0].legend(fontsize=8); axes[0].grid(True, alpha=0.3)

axes[1].plot(epochs, rs, "o-", color="forestgreen")
axes[1].axhline(0.2887, color="purple", lw=1.2,
                linestyle=":", label="CNN+LSTM r=0.289")
axes[1].set_title("Pearson r (predicted vs actual)")
axes[1].set_xlabel("Epoch"); axes[1].set_ylabel("Pearson r")
axes[1].set_ylim(0, 1); axes[1].legend(fontsize=8)
axes[1].grid(True, alpha=0.3)

sample_idx = np.random.choice(min(len(pf), 500000), 5000, replace=False)
axes[2].scatter(ff[:500000][sample_idx], pf[:500000][sample_idx],
                alpha=0.1, s=1, color="steelblue")
axes[2].plot([-1,1], [-1,1], "r--", lw=1.5, label="Perfect")
axes[2].set_xlabel("Actual R_ewma"); axes[2].set_ylabel("Predicted R_ewma")
axes[2].set_title(f"Predicted vs Actual (r={pearson_r_final:.3f})")
axes[2].legend(fontsize=9); axes[2].grid(True, alpha=0.3)

plt.tight_layout()
path = os.path.join(PROJECT_DIR, "plot_cell14_3dcnn_adaptive.png")
plt.savefig(path, dpi=150, bbox_inches="tight")
plt.show()
print(f"âœ“ Saved: {path}")
print("\nâœ“ Cell 14 complete. Run Cell 15 for ConvLSTM.")
```

```python
# Cell 15 â€” ConvLSTM Regression with OneCycleLR
#
# ConvLSTM: the hidden state is a FEATURE MAP, not a vector.
# Unlike CNN+LSTM where spatial structure is lost after CNN encoding,
# ConvLSTM carries the full spatial map through time.
# At each timestep, convolutional gates update the spatial hidden state â€”
# so the model tracks how each region of the correlation matrix evolves.
#
# OneCycleLR: actively drives LR through explore â†’ converge â†’ fine-tune.
# Instead of passively waiting for plateau and cutting,
# LR increases first (escape local minima, overshoot intentionally)
# then decreases smoothly (settle into a better minimum).
# Proposed by Leslie Smith (2017), popularized by fast.ai.

import numpy as np
import pickle, os, time
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from tqdm import tqdm
from scipy.stats import pearsonr
import matplotlib.pyplot as plt

PROJECT_DIR = r"C:\Users\HP\Documents\InterStockRelation_Paper\InterStockRelation"
MODELS_DIR  = os.path.join(PROJECT_DIR, "models")

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Device : {device}  ({torch.cuda.get_device_name(0)})")

# â”€â”€ Load data â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
print("\nLoading data...")
corr_video = np.load(os.path.join(PROJECT_DIR, "corr_video.npy"))

with open(os.path.join(PROJECT_DIR, "dataset_meta_v2.pkl"), "rb") as f:
    dm = pickle.load(f)

N_STOCKS            = dm["n_stocks"]
HORIZON             = dm["horizon"]
LOOKBACK            = dm["lookback"]
STRIDE_DAILY        = dm["stride_daily"]
STRIDE_SUBDAILY     = dm["stride_subdaily"]
daily_train         = dm["daily_train"]
daily_val           = dm["daily_val"]
subdaily_train      = dm["subdaily_train"]
subdaily_frame_cache= dm["subdaily_frame_cache"]
frame_mse           = dm["frame_mse"]
persistence_mse     = dm["persistence_mse"]
mean_frame_mse      = dm["mean_frame_mse"]
volatility_threshold= dm["volatility_threshold"]
snap_dates          = dm["snap_dates"]

print(f"âœ“ corr_video      : {corr_video.shape}")
print(f"  Daily train     : {len(daily_train)}")
print(f"  Subdaily train  : {len(subdaily_train)}")
print(f"  Val snaps       : {len(daily_val)}")
print(f"  Persistence MSE : {persistence_mse:.6f}")
print(f"  Mean-frame MSE  : {mean_frame_mse:.6f}")

# â”€â”€ Dataset â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
class AdaptiveResDataset(Dataset):
    def __init__(self, daily_snaps, subdaily_snaps=None):
        self.samples = [("daily", s) for s in daily_snaps]
        if subdaily_snaps:
            self.samples += [("subdaily", t) for t in subdaily_snaps]

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        kind, index = self.samples[idx]
        if kind == "daily":
            frames = corr_video[index - LOOKBACK: index, :, :]
            future = corr_video[index + HORIZON]
            weight = float(frame_mse[index])
        else:
            t_idx  = index
            frames = np.stack([
                subdaily_frame_cache.get(
                    t_idx - (LOOKBACK - k) * STRIDE_SUBDAILY,
                    np.zeros((N_STOCKS, N_STOCKS), dtype=np.float32))
                for k in range(LOOKBACK)
            ], axis=0)
            t_future = t_idx + HORIZON * STRIDE_DAILY
            future   = subdaily_frame_cache.get(
                t_future, np.zeros((N_STOCKS, N_STOCKS), dtype=np.float32))
            weight   = float(np.mean((frames[-1] - future) ** 2))
        return (
            torch.from_numpy(frames.copy()).float(),
            torch.from_numpy(np.array(future).copy()).float(),
            torch.tensor(weight, dtype=torch.float32),
        )

BATCH_SIZE = 16
train_ds   = AdaptiveResDataset(daily_train, subdaily_train)
val_ds     = AdaptiveResDataset(daily_val,   None)
train_dl   = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True,
                        num_workers=0, pin_memory=True)
val_dl     = DataLoader(val_ds,   batch_size=BATCH_SIZE, shuffle=False,
                        num_workers=0, pin_memory=True)

print(f"\nTrain samples : {len(train_ds)}")
print(f"Val samples   : {len(val_ds)}")

# â”€â”€ ConvLSTM Cell â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
class ConvLSTMCell(nn.Module):
    """
    Single ConvLSTM cell.
    Hidden state h and cell state c are spatial feature maps,
    not vectors â€” spatial structure is preserved through time.

    Input  : (B, input_channels, H, W)
    Output : h, c both (B, hidden_channels, H, W)
    """
    def __init__(self, input_channels, hidden_channels, kernel_size=3):
        super().__init__()
        pad = kernel_size // 2
        # All four gates computed in one convolution for efficiency
        self.conv = nn.Conv2d(
            input_channels + hidden_channels,
            4 * hidden_channels,   # i, f, g, o gates
            kernel_size=kernel_size,
            padding=pad
        )
        self.hidden_channels = hidden_channels

    def forward(self, x, h, c):
        # x: (B, input_ch, H, W)
        # h: (B, hidden_ch, H, W)
        combined = torch.cat([x, h], dim=1)         # (B, input+hidden, H, W)
        gates    = self.conv(combined)               # (B, 4*hidden, H, W)

        # Split into four gates
        i, f, g, o = gates.chunk(4, dim=1)
        i = torch.sigmoid(i)   # input gate
        f = torch.sigmoid(f)   # forget gate
        g = torch.tanh(g)      # cell gate
        o = torch.sigmoid(o)   # output gate

        c_next = f * c + i * g
        h_next = o * torch.tanh(c_next)
        return h_next, c_next

    def init_hidden(self, batch_size, spatial_size, device):
        H, W = spatial_size
        h = torch.zeros(batch_size, self.hidden_channels, H, W, device=device)
        c = torch.zeros(batch_size, self.hidden_channels, H, W, device=device)
        return h, c


class ConvLSTMReg(nn.Module):
    """
    ConvLSTM regression model for correlation matrix video.

    Architecture:
      1. Spatial downsampler: reduce 95Ã—95 to manageable spatial size
         before ConvLSTM (saves memory, speeds up training)
      2. Two ConvLSTM layers: process the sequence of downsampled frames
         Hidden state carries spatial cluster structure through time
      3. Spatial upsampler + decoder: reconstruct 95Ã—95 future matrix

    Input : (B, LOOKBACK, 95, 95)
    Output: (B, 95, 95)
    """
    def __init__(self, n_stocks=95, spatial_size=24,
                 hidden_channels=32, dropout=0.3):
        super().__init__()
        self.n_stocks    = n_stocks
        self.spatial_size= spatial_size
        self.hidden_ch   = hidden_channels

        # Downsample 95Ã—95 â†’ spatial_sizeÃ—spatial_size
        # Keeps spatial structure but reduces memory
        self.downsample = nn.Sequential(
            nn.Conv2d(1,  16, kernel_size=3, padding=1),
            nn.BatchNorm2d(16), nn.ReLU(),
            nn.Conv2d(16, 32, kernel_size=3, padding=1),
            nn.BatchNorm2d(32), nn.ReLU(),
            nn.AdaptiveAvgPool2d((spatial_size, spatial_size)),
            nn.Conv2d(32, hidden_channels, kernel_size=1),
        )

        # Two ConvLSTM layers
        self.convlstm1 = ConvLSTMCell(hidden_channels, hidden_channels)
        self.convlstm2 = ConvLSTMCell(hidden_channels, hidden_channels)
        self.layer_norm= nn.LayerNorm([hidden_channels, spatial_size, spatial_size])
        self.dropout2d = nn.Dropout2d(dropout)

        # Decoder: spatial map â†’ future correlation matrix
        # Flatten spatial hidden state â†’ predict 95Ã—95
        flat_size = hidden_channels * spatial_size * spatial_size
        self.decoder = nn.Sequential(
            nn.Flatten(),
            nn.Linear(flat_size, 1024),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(1024, 2048),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(2048, n_stocks * n_stocks),
            nn.Tanh()
        )

    def forward(self, frames):
        """frames: (B, LOOKBACK, 95, 95)"""
        B, L, H, W = frames.shape
        sp         = self.spatial_size

        # Encode each frame spatially
        frames_flat = frames.reshape(B * L, 1, H, W)
        encoded     = self.downsample(frames_flat)         # (B*L, hidden, sp, sp)
        encoded_seq = encoded.reshape(B, L, self.hidden_ch, sp, sp)

        # Init hidden states
        h1, c1 = self.convlstm1.init_hidden(B, (sp, sp), frames.device)
        h2, c2 = self.convlstm2.init_hidden(B, (sp, sp), frames.device)

        # Unroll ConvLSTM across time
        for t in range(L):
            x_t    = encoded_seq[:, t, :, :, :]           # (B, hidden, sp, sp)
            h1, c1 = self.convlstm1(x_t, h1, c1)
            h1     = self.dropout2d(h1)
            h2, c2 = self.convlstm2(h1, h2, c2)

        # Final hidden state h2: (B, hidden, sp, sp)
        h2  = self.layer_norm(h2)

        # Decode to future correlation matrix
        out = self.decoder(h2)                             # (B, 95*95)
        return out.reshape(B, self.n_stocks, self.n_stocks)


def weighted_mse(pred, target, weights):
    per_sample = ((pred - target) ** 2).mean(dim=(1, 2))
    w = weights / weights.sum() * len(weights)
    return (w * per_sample).mean()

# â”€â”€ Training config â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
MAX_EPOCHS = 40
BASE_LR    = 1e-3
PATIENCE   = 8

# â”€â”€ Custom directional LR scheduler â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
class DirectionalLRScheduler:
    """
    Adjusts LR based on direction of progress toward the target.
    On bad steps: decrease LR AND rewind weights to best checkpoint.
    On good steps: increase LR, keep current weights.

    Rewind counter prevents infinite loops around the same checkpoint â€”
    after MAX_REWINDS bad steps from the same checkpoint, accept current
    weights and continue exploring from there.
    """
    def __init__(self, optimizer, model, min_lr=1e-5, max_lr=1e-2,
                 increase_factor=0.15, decrease_factor=0.30,
                 smooth_window=5, max_rewinds=3):
        self.optimizer        = optimizer
        self.model            = model
        self.min_lr           = min_lr
        self.max_lr           = max_lr
        self.increase_factor  = increase_factor
        self.decrease_factor  = decrease_factor
        self.smooth_window    = smooth_window
        self.max_rewinds      = max_rewinds
        self.vs_mean_history  = []
        self.pearson_history  = []
        self.lr_history       = []
        self.best_state       = None   # cached best model weights
        self.best_vs_mean     = -float("inf")
        self.rewind_count     = 0     # how many times rewound from current best

    def get_lr(self):
        return self.optimizer.param_groups[0]["lr"]

    def cache_best(self, vs_mean):
        """Cache current model weights as new best."""
        import copy
        self.best_state   = copy.deepcopy(self.model.state_dict())
        self.best_vs_mean = vs_mean
        self.rewind_count = 0   # reset rewind counter for new best

    def rewind(self):
        """Restore model weights to best cached state."""
        if self.best_state is not None:
            self.model.load_state_dict(self.best_state)
            self.rewind_count += 1
            return True
        return False

    def step(self, vs_mean, pearson_r):
        self.vs_mean_history.append(vs_mean)
        self.pearson_history.append(pearson_r)

        current_lr = self.get_lr()

        # Cache if this is the best vs_mean seen so far
        if vs_mean > self.best_vs_mean:
            self.cache_best(vs_mean)

        # Need enough history for smoothing
        if len(self.vs_mean_history) < self.smooth_window + 1:
            self.lr_history.append(current_lr)
            return current_lr, "warmup", False

        # Smoothed signals
        smooth_vm_now  = np.mean(self.vs_mean_history[-self.smooth_window:])
        smooth_vm_prev = np.mean(self.vs_mean_history[-self.smooth_window-1:-1])
        smooth_r_now   = np.mean(self.pearson_history[-self.smooth_window:])
        smooth_r_prev  = np.mean(self.pearson_history[-self.smooth_window-1:-1])

        delta_vm = smooth_vm_now - smooth_vm_prev
        delta_r  = smooth_r_now  - smooth_r_prev

        vm_improving = delta_vm > 0
        r_improving  = delta_r  > 0
        rewound      = False

        if vm_improving and r_improving:
            # Both improving â€” reward with LR increase
            proximity = max(0.1, min(1.0 - abs(smooth_vm_now) / 100.0, 1.0))
            new_lr    = current_lr * (1.0 + self.increase_factor * proximity)
            new_lr    = min(new_lr, self.max_lr)
            direction = f"â†‘ both improving (vm={delta_vm:+.2f} r={delta_r:+.4f})"

        elif vm_improving and not r_improving:
            # vs_mean improving but Pearson r not â€” cautious small increase
            new_lr    = current_lr * 1.03
            new_lr    = min(new_lr, self.max_lr)
            direction = f"â†’ vm only (vm={delta_vm:+.2f} r={delta_r:+.4f})"

        else:
            # Wrong direction â€” decrease LR
            new_lr = current_lr * (1.0 - self.decrease_factor)
            new_lr = max(new_lr, self.min_lr)

            # Rewind to best weights if we haven't exhausted rewind budget
            if self.rewind_count < self.max_rewinds:
                did_rewind = self.rewind()
                rewind_str = f" [rewind {self.rewind_count}/{self.max_rewinds}]" \
                             if did_rewind else ""
                rewound    = did_rewind
            else:
                rewind_str = " [rewind budget exhausted â€” accepting current]"
                self.rewind_count = 0   # reset so future bads can rewind again

            direction = (f"â†“ off track (vm={delta_vm:+.2f} "
                         f"r={delta_r:+.4f}){rewind_str}")

        for pg in self.optimizer.param_groups:
            pg["lr"] = new_lr

        self.lr_history.append(new_lr)
        return new_lr, direction, rewound

torch.cuda.empty_cache()
model     = ConvLSTMReg(n_stocks=N_STOCKS).to(device)
optimizer = torch.optim.Adam(model.parameters(), lr=BASE_LR, weight_decay=1e-4)
scheduler = DirectionalLRScheduler(
    optimizer,
    model           = model,
    min_lr          = 1e-5,
    max_lr          = 1e-2,
    increase_factor = 0.15,
    decrease_factor = 0.30,
    smooth_window   = 5,
    max_rewinds     = 3,
)

total_params = sum(p.numel() for p in model.parameters())
print(f"\nParameters : {total_params:,}")
print(f"Baselines  : persistence={persistence_mse:.4f}  "
      f"mean-frame={mean_frame_mse:.4f}")
print(f"OneCycleLR : base={BASE_LR:.0e} â†’ peak={MAX_LR:.0e} â†’ ~{MAX_LR/1e4:.0e}")
print()

best_val_mse = float("inf")
patience_ctr = 0
best_path    = os.path.join(MODELS_DIR, "convlstm_reg_best.pt")
history      = []

for epoch in range(1, MAX_EPOCHS + 1):
    t0 = time.time()

    # â”€â”€ Train â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
    model.train()
    tr_loss = tr_total = 0.0

    for frames, future, w in tqdm(train_dl,
            desc=f"Ep {epoch:>2} Train", ncols=80, leave=False):
        frames, future, w = frames.to(device), future.to(device), w.to(device)
        optimizer.zero_grad()
        pred = model(frames)
        loss = weighted_mse(pred, future, w)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        # scheduler.step() removed â€” directional scheduler is per-epoch
        tr_loss  += loss.item() * frames.size(0)
        tr_total += frames.size(0)

    # â”€â”€ Validate â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
    model.eval()
    va_loss = va_total = 0.0
    all_preds, all_futures, all_weights = [], [], []

    with torch.no_grad():
        for frames, future, w in tqdm(val_dl,
                desc=f"Ep {epoch:>2} Val  ", ncols=80, leave=False):
            frames, future, w = frames.to(device), future.to(device), w.to(device)
            pred     = model(frames)
            loss     = weighted_mse(pred, future, w)
            va_loss += loss.item() * frames.size(0)
            va_total+= frames.size(0)
            all_preds.append(pred.cpu().numpy())
            all_futures.append(future.cpu().numpy())
            all_weights.extend(w.cpu().numpy())

    tr_mse   = tr_loss / tr_total
    va_mse   = va_loss / va_total
    lr_now   = optimizer.param_groups[0]["lr"]
    elapsed  = time.time() - t0

    pf = np.concatenate(all_preds).reshape(-1)
    ff = np.concatenate(all_futures).reshape(-1)
    pearson_r, _ = pearsonr(pf[:100000], ff[:100000])

    vs_persist = (persistence_mse - va_mse) / persistence_mse * 100
    vs_mean    = (mean_frame_mse  - va_mse) / mean_frame_mse  * 100
    new_lr, direction, rewound = scheduler.step(vs_mean, pearson_r)
    improved   = " âœ“" if va_mse < best_val_mse else ""

    history.append(dict(epoch=epoch, tr_mse=tr_mse, va_mse=va_mse,
                        pearson_r=pearson_r, lr=new_lr))

    print(f"Ep {epoch:>2}/{MAX_EPOCHS} | "
          f"tr={tr_mse:.5f} | va={va_mse:.5f} | "
          f"r={pearson_r:.4f} | "
          f"vs_persist={vs_persist:+.1f}% | "
          f"vs_mean={vs_mean:+.1f}% | "
          f"lr={new_lr:.2e} | {direction} | {elapsed:.0f}s{improved}")

    if va_mse < best_val_mse:
        best_val_mse = va_mse
        patience_ctr = 0
        torch.save(model.state_dict(), best_path)
    else:
        patience_ctr += 1
        if patience_ctr >= PATIENCE:
            print(f"\nEarly stopping at epoch {epoch}")
            break

print(f"\nâœ“ Best val MSE : {best_val_mse:.6f}")
print(f"  vs persistence : {(persistence_mse-best_val_mse)/persistence_mse*100:+.1f}%")
print(f"  vs mean-frame  : {(mean_frame_mse-best_val_mse)/mean_frame_mse*100:+.1f}%")

# â”€â”€ Final evaluation â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
model.load_state_dict(torch.load(best_path, map_location=device))
model.eval()
all_preds, all_futures, all_weights = [], [], []

with torch.no_grad():
    for frames, future, w in val_dl:
        pred = model(frames.to(device)).cpu().numpy()
        all_preds.append(pred)
        all_futures.append(future.numpy())
        all_weights.extend(w.numpy())

all_preds    = np.concatenate(all_preds,   axis=0)
all_futures  = np.concatenate(all_futures, axis=0)
all_weights  = np.array(all_weights)
per_snap_mse = ((all_preds - all_futures) ** 2).mean(axis=(1, 2))

val_current  = np.array([corr_video[s] for s in daily_val[:len(all_preds)]])
dir_acc      = (np.sign(all_preds - val_current) ==
                np.sign(all_futures - val_current)).mean()

pf = all_preds.reshape(-1);  ff = all_futures.reshape(-1)
pearson_r_final, _ = pearsonr(pf[:500000], ff[:500000])

volatile_mask = all_weights >= volatility_threshold
mse_volatile  = per_snap_mse[volatile_mask].mean() if volatile_mask.any() else float("nan")
mse_stable    = per_snap_mse[~volatile_mask].mean() if (~volatile_mask).any() else float("nan")

print(f"\nâ”€â”€ Final Evaluation â”€â”€")
print(f"  Val MSE (unweighted)  : {per_snap_mse.mean():.6f}  (CNN+LSTM: 0.072454)")
print(f"  Persistence MSE       : {persistence_mse:.6f}")
print(f"  Mean-frame MSE        : {mean_frame_mse:.6f}")
print(f"  Pearson r             : {pearson_r_final:.4f}  (CNN+LSTM: 0.2887)")
print(f"  Directional accuracy  : {dir_acc*100:.2f}%  (CNN+LSTM: 73.03%)")
print(f"  MSE on volatile snaps : {mse_volatile:.6f}  (CNN+LSTM: 0.109566)")
print(f"  MSE on stable snaps   : {mse_stable:.6f}  (CNN+LSTM: 0.036622)")

# â”€â”€ Save â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
np.save(os.path.join(PROJECT_DIR, "convlstm_val_preds.npy"), all_preds)
with open(os.path.join(MODELS_DIR, "convlstm_reg_history.pkl"), "wb") as f:
    pickle.dump(history, f)
print(f"\nâœ“ Saved: {best_path}")

# â”€â”€ Release GPU â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
del model
torch.cuda.empty_cache()
print(f"âœ“ GPU memory released")

# â”€â”€ Plot â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
fig, axes = plt.subplots(1, 4, figsize=(22, 5))
fig.suptitle("ConvLSTM Regression â€” Adaptive Resolution + OneCycleLR",
             fontsize=13, fontweight="bold")

epochs  = [h["epoch"]     for h in history]
tr_mses = [h["tr_mse"]    for h in history]
va_mses = [h["va_mse"]    for h in history]
rs      = [h["pearson_r"] for h in history]
lrs     = [h["lr"]        for h in history]

axes[0].plot(epochs, tr_mses, "o-", color="steelblue",  label="Train MSE")
axes[0].plot(epochs, va_mses, "o-", color="darkorange", label="Val MSE")
axes[0].axhline(persistence_mse, color="red",    lw=1.5,
                linestyle="--", label=f"Persistence {persistence_mse:.4f}")
axes[0].axhline(mean_frame_mse,  color="green",  lw=1.5,
                linestyle="--", label=f"Mean-frame {mean_frame_mse:.4f}")
axes[0].axhline(0.075715, color="purple", lw=1.2,
                linestyle=":", label="CNN+LSTM 0.0757")
axes[0].set_title("MSE over epochs")
axes[0].set_xlabel("Epoch"); axes[0].set_ylabel("MSE")
axes[0].legend(fontsize=7); axes[0].grid(True, alpha=0.3)

axes[1].plot(epochs, rs, "o-", color="forestgreen")
axes[1].axhline(0.2887, color="purple", lw=1.2,
                linestyle=":", label="CNN+LSTM r=0.289")
axes[1].set_title("Pearson r"); axes[1].set_xlabel("Epoch")
axes[1].set_ylabel("Pearson r"); axes[1].set_ylim(0, 1)
axes[1].legend(fontsize=8); axes[1].grid(True, alpha=0.3)

# LR schedule â€” shows the OneCycleLR curve
axes[2].plot(epochs, lrs, "o-", color="crimson")
axes[2].set_title("Learning Rate (OneCycleLR)")
axes[2].set_xlabel("Epoch"); axes[2].set_ylabel("LR")
axes[2].set_yscale("log"); axes[2].grid(True, alpha=0.3)

sample_idx = np.random.choice(min(len(pf), 500000), 5000, replace=False)
axes[3].scatter(ff[:500000][sample_idx], pf[:500000][sample_idx],
                alpha=0.1, s=1, color="steelblue")
axes[3].plot([-1,1], [-1,1], "r--", lw=1.5, label="Perfect")
axes[3].set_xlabel("Actual R_ewma"); axes[3].set_ylabel("Predicted R_ewma")
axes[3].set_title(f"Predicted vs Actual (r={pearson_r_final:.3f})")
axes[3].legend(fontsize=9); axes[3].grid(True, alpha=0.3)

plt.tight_layout()
path = os.path.join(PROJECT_DIR, "plot_cell15_convlstm.png")
plt.savefig(path, dpi=150, bbox_inches="tight")
plt.show()
print(f"âœ“ Saved: {path}")
print("\nâœ“ Cell 15 complete. Run Cell 16 for ensemble.")
```

```python
# Cell 16 â€” Pretrained EfficientNet-B0 + LSTM (Transfer Learning)
#
# Uses EfficientNet-B0 pretrained on ImageNet as the per-frame spatial encoder.
# Replaces our hand-built FrameCNN with a much stronger feature extractor.
# The LSTM then processes the sequence of EfficientNet features across frames.
#
# Transfer learning strategy:
#   - Freeze EfficientNet backbone initially (use pretrained features as-is)
#   - Only train the LSTM + decoder for first FREEZE_EPOCHS epochs
#   - Then unfreeze last N EfficientNet blocks for fine-tuning
#   - This prevents destroying pretrained features before LSTM has stabilized
#
# Input adaptation:
#   - Our frames are 1-channel 95Ã—95
#   - EfficientNet expects 3-channel 224Ã—224
#   - We repeat the channel 3 times and upsample to 224Ã—224
#   - AdaptiveAvgPool in EfficientNet handles size automatically
#
# Directional LR scheduler with weight rewind (from Cell 15)
# Adaptive resolution dataset (from Cell 12b)

import numpy as np
import pickle, os, time, copy
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import models
from tqdm import tqdm
from scipy.stats import pearsonr
import matplotlib.pyplot as plt

PROJECT_DIR = r"C:\Users\HP\Documents\InterStockRelation_Paper\InterStockRelation"
MODELS_DIR  = os.path.join(PROJECT_DIR, "models")

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Device : {device}  ({torch.cuda.get_device_name(0)})")

# â”€â”€ Load data â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
print("\nLoading data...")
corr_video = np.load(os.path.join(PROJECT_DIR, "corr_video.npy"))

with open(os.path.join(PROJECT_DIR, "dataset_meta_v2.pkl"), "rb") as f:
    dm = pickle.load(f)

N_STOCKS            = dm["n_stocks"]
HORIZON             = dm["horizon"]
LOOKBACK            = dm["lookback"]
STRIDE_DAILY        = dm["stride_daily"]
STRIDE_SUBDAILY     = dm["stride_subdaily"]
daily_train         = dm["daily_train"]
daily_val           = dm["daily_val"]
subdaily_train      = dm["subdaily_train"]
subdaily_frame_cache= dm["subdaily_frame_cache"]
frame_mse           = dm["frame_mse"]
persistence_mse     = dm["persistence_mse"]
mean_frame_mse      = dm["mean_frame_mse"]
volatility_threshold= dm["volatility_threshold"]
snap_dates          = dm["snap_dates"]

print(f"âœ“ corr_video      : {corr_video.shape}")
print(f"  Daily train     : {len(daily_train)}")
print(f"  Subdaily train  : {len(subdaily_train)}")
print(f"  Val snaps       : {len(daily_val)}")
print(f"  Persistence MSE : {persistence_mse:.6f}")
print(f"  Mean-frame MSE  : {mean_frame_mse:.6f}")

# â”€â”€ Dataset â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
class AdaptiveResDataset(Dataset):
    def __init__(self, daily_snaps, subdaily_snaps=None):
        self.samples = [("daily", s) for s in daily_snaps]
        if subdaily_snaps:
            self.samples += [("subdaily", t) for t in subdaily_snaps]

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        kind, index = self.samples[idx]
        if kind == "daily":
            frames = corr_video[index - LOOKBACK: index, :, :]
            future = corr_video[index + HORIZON]
            weight = float(frame_mse[index])
        else:
            t_idx  = index
            frames = np.stack([
                subdaily_frame_cache.get(
                    t_idx - (LOOKBACK - k) * STRIDE_SUBDAILY,
                    np.zeros((N_STOCKS, N_STOCKS), dtype=np.float32))
                for k in range(LOOKBACK)
            ], axis=0)
            t_future = t_idx + HORIZON * STRIDE_DAILY
            future   = subdaily_frame_cache.get(
                t_future, np.zeros((N_STOCKS, N_STOCKS), dtype=np.float32))
            weight   = float(np.mean((frames[-1] - future) ** 2))
        return (
            torch.from_numpy(frames.copy()).float(),
            torch.from_numpy(np.array(future).copy()).float(),
            torch.tensor(weight, dtype=torch.float32),
        )

# Smaller batch â€” EfficientNet + LSTM is memory heavy
BATCH_SIZE = 8
train_ds   = AdaptiveResDataset(daily_train, subdaily_train)
val_ds     = AdaptiveResDataset(daily_val,   None)
train_dl   = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True,
                        num_workers=0, pin_memory=True)
val_dl     = DataLoader(val_ds,   batch_size=BATCH_SIZE, shuffle=False,
                        num_workers=0, pin_memory=True)

print(f"\nTrain samples : {len(train_ds)}")
print(f"Val samples   : {len(val_ds)}")

# â”€â”€ Architecture â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€

class EfficientNetEncoder(nn.Module):
    """
    EfficientNet-B0 pretrained on ImageNet as spatial frame encoder.
    Input : (B, 1, 95, 95) â€” single grayscale correlation frame
    Output: (B, feature_dim) â€” spatial feature vector

    Adaptation:
      1. Repeat 1-channel input to 3 channels (fake RGB)
      2. EfficientNet extracts spatial features
      3. Global average pool to fixed-size vector
    """
    def __init__(self, freeze_backbone=True):
        super().__init__()

        # Load pretrained EfficientNet-B0
        efficientnet = models.efficientnet_b0(
            weights=models.EfficientNet_B0_Weights.IMAGENET1K_V1)

        # Remove the classification head â€” keep only the feature extractor
        self.backbone = efficientnet.features   # convolutional feature extractor
        self.pool     = efficientnet.avgpool    # global average pooling
        # EfficientNet-B0 outputs 1280-dim features after pooling
        self.feature_dim = 1280

        # Layer norm for stability
        self.norm = nn.LayerNorm(self.feature_dim)

        if freeze_backbone:
            for param in self.backbone.parameters():
                param.requires_grad = False
            print("âœ“ EfficientNet backbone frozen (pretrained features)")
        else:
            print("âœ“ EfficientNet backbone unfrozen (fine-tuning)")

    def unfreeze_last_n_blocks(self, n=3):
        """Unfreeze last n blocks of EfficientNet for fine-tuning."""
        blocks = list(self.backbone.children())
        for block in blocks[-n:]:
            for param in block.parameters():
                param.requires_grad = True
        n_trainable = sum(p.numel() for p in self.parameters()
                         if p.requires_grad)
        print(f"âœ“ Unfroze last {n} blocks â€” {n_trainable:,} trainable params")

    def forward(self, x):
        # x: (B, 1, 95, 95)
        # Repeat channel to fake RGB: (B, 1, 95, 95) â†’ (B, 3, 95, 95)
        x = x.repeat(1, 3, 1, 1)
        # Upsample to 224Ã—224 for EfficientNet
        x = F.interpolate(x, size=(224, 224), mode="bilinear",
                          align_corners=False)
        # Extract features
        x = self.backbone(x)               # (B, 1280, 7, 7)
        x = self.pool(x)                   # (B, 1280, 1, 1)
        x = x.flatten(1)                   # (B, 1280)
        return self.norm(x)


class EfficientNetLSTMReg(nn.Module):
    """
    EfficientNet-B0 frame encoder + LSTM temporal model.

    Input : (B, LOOKBACK, 95, 95)
    Output: (B, 95, 95) â€” predicted future correlation matrix

    Architecture:
      1. EfficientNetEncoder: each frame â†’ 1280-dim spatial features
      2. Projection: 1280 â†’ lstm_input_dim (reduce before LSTM)
      3. LSTM: sequence of frame features â†’ temporal understanding
      4. Decoder: LSTM hidden â†’ 95Ã—95 future matrix
    """
    def __init__(self, n_stocks=95, lstm_hidden=256, lstm_layers=2,
                 lstm_input_dim=256, dropout=0.3, freeze_backbone=True):
        super().__init__()
        self.n_stocks  = n_stocks
        self.encoder   = EfficientNetEncoder(freeze_backbone)

        # Project 1280-dim EfficientNet features to LSTM input size
        self.proj = nn.Sequential(
            nn.Linear(self.encoder.feature_dim, lstm_input_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.LayerNorm(lstm_input_dim),
        )

        self.lstm    = nn.LSTM(lstm_input_dim, lstm_hidden, lstm_layers,
                               batch_first=True,
                               dropout=dropout if lstm_layers > 1 else 0.0)
        self.norm    = nn.LayerNorm(lstm_hidden)
        self.dropout = nn.Dropout(dropout)

        self.decoder = nn.Sequential(
            nn.Linear(lstm_hidden, 512),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(512, 1024),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(1024, n_stocks * n_stocks),
            nn.Tanh()
        )

    def forward(self, frames):
        B, L, H, W  = frames.shape
        frames_flat = frames.reshape(B * L, 1, H, W)

        # Encode each frame with EfficientNet
        enc_flat    = self.encoder(frames_flat)           # (B*L, 1280)
        enc_proj    = self.proj(enc_flat)                 # (B*L, lstm_input_dim)
        enc_seq     = enc_proj.reshape(B, L, -1)         # (B, L, lstm_input_dim)

        # LSTM across frames
        lstm_out, _ = self.lstm(enc_seq)
        h           = self.norm(lstm_out[:, -1, :])      # (B, lstm_hidden)
        h           = self.dropout(h)

        out = self.decoder(h)                             # (B, 95*95)
        return out.reshape(B, self.n_stocks, self.n_stocks)

    def unfreeze_backbone(self, n_blocks=3):
        self.encoder.unfreeze_last_n_blocks(n_blocks)


# â”€â”€ Directional LR Scheduler with weight rewind (from Cell 15) â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
class DirectionalLRScheduler:
    def __init__(self, optimizer, model, min_lr=1e-5, max_lr=5e-3,
                 increase_factor=0.15, decrease_factor=0.30,
                 smooth_window=3, max_rewinds=3):
        self.optimizer        = optimizer
        self.model            = model
        self.min_lr           = min_lr
        self.max_lr           = max_lr
        self.increase_factor  = increase_factor
        self.decrease_factor  = decrease_factor
        self.smooth_window    = smooth_window
        self.max_rewinds      = max_rewinds
        self.vs_mean_history  = []
        self.pearson_history  = []
        self.lr_history       = []
        self.best_state       = None
        self.best_vs_mean     = -float("inf")
        self.rewind_count     = 0

    def get_lr(self):
        return self.optimizer.param_groups[0]["lr"]

    def cache_best(self, vs_mean):
        self.best_state   = copy.deepcopy(self.model.state_dict())
        self.best_vs_mean = vs_mean
        self.rewind_count = 0

    def rewind(self):
        if self.best_state is not None:
            self.model.load_state_dict(self.best_state)
            self.rewind_count += 1
            return True
        return False

    def step(self, vs_mean, pearson_r):
        self.vs_mean_history.append(vs_mean)
        self.pearson_history.append(pearson_r)
        current_lr = self.get_lr()

        if vs_mean > self.best_vs_mean:
            self.cache_best(vs_mean)

        if len(self.vs_mean_history) < self.smooth_window + 1:
            self.lr_history.append(current_lr)
            return current_lr, "warmup", False

        smooth_vm_now  = np.mean(self.vs_mean_history[-self.smooth_window:])
        smooth_vm_prev = np.mean(self.vs_mean_history[-self.smooth_window-1:-1])
        smooth_r_now   = np.mean(self.pearson_history[-self.smooth_window:])
        smooth_r_prev  = np.mean(self.pearson_history[-self.smooth_window-1:-1])

        delta_vm = smooth_vm_now - smooth_vm_prev
        delta_r  = smooth_r_now  - smooth_r_prev
        rewound  = False

        if delta_vm > 0 and delta_r > 0:
            proximity = max(0.1, min(1.0 - abs(smooth_vm_now) / 100.0, 1.0))
            new_lr    = min(current_lr * (1.0 + self.increase_factor * proximity),
                           self.max_lr)
            direction = f"â†‘ both improving (vm={delta_vm:+.2f} r={delta_r:+.4f})"
        elif delta_vm > 0:
            new_lr    = min(current_lr * 1.03, self.max_lr)
            direction = f"â†’ vm only (vm={delta_vm:+.2f} r={delta_r:+.4f})"
        else:
            new_lr = max(current_lr * (1.0 - self.decrease_factor), self.min_lr)
            if self.rewind_count < self.max_rewinds:
                rewound    = self.rewind()
                rewind_str = f" [rewind {self.rewind_count}/{self.max_rewinds}]"
            else:
                rewind_str        = " [budget exhausted â€” exploring forward]"
                self.rewind_count = 0
            direction = (f"â†“ off track (vm={delta_vm:+.2f} "
                        f"r={delta_r:+.4f}){rewind_str}")

        for pg in self.optimizer.param_groups:
            pg["lr"] = new_lr
        self.lr_history.append(new_lr)
        return new_lr, direction, rewound


# â”€â”€ Training config â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
MAX_EPOCHS    = 40
BASE_LR       = 5e-4    # lower than before â€” EfficientNet features are sensitive
PATIENCE      = 10
FREEZE_EPOCHS = 5       # train only LSTM+decoder first, then unfreeze backbone

def weighted_mse(pred, target, weights):
    per_sample = ((pred - target) ** 2).mean(dim=(1, 2))
    w = weights / weights.sum() * len(weights)
    return (w * per_sample).mean()

torch.cuda.empty_cache()
model     = EfficientNetLSTMReg(n_stocks=N_STOCKS, freeze_backbone=True).to(device)
optimizer = torch.optim.Adam(
    filter(lambda p: p.requires_grad, model.parameters()),
    lr=BASE_LR, weight_decay=1e-4)
scheduler = DirectionalLRScheduler(
    optimizer, model,
    min_lr=1e-6, max_lr=5e-3,
    increase_factor=0.15, decrease_factor=0.30,
    smooth_window=3, max_rewinds=3,
)

frozen_params   = sum(p.numel() for p in model.parameters()
                      if not p.requires_grad)
trainable_params= sum(p.numel() for p in model.parameters()
                      if p.requires_grad)
print(f"\nFrozen params    : {frozen_params:,}  (EfficientNet backbone)")
print(f"Trainable params : {trainable_params:,}  (LSTM + decoder + projection)")
print(f"Baselines        : persistence={persistence_mse:.4f}  "
      f"mean-frame={mean_frame_mse:.4f}")
print(f"Freeze epochs    : {FREEZE_EPOCHS} â†’ then unfreeze last 3 blocks\n")

best_val_mse  = float("inf")
patience_ctr  = 0
best_path     = os.path.join(MODELS_DIR, "efficientnet_lstm_best.pt")
history       = []
backbone_unfrozen = False

for epoch in range(1, MAX_EPOCHS + 1):
    t0 = time.time()

    # Unfreeze backbone after FREEZE_EPOCHS
    if epoch == FREEZE_EPOCHS + 1 and not backbone_unfrozen:
        model.unfreeze_backbone(n_blocks=3)
        # Rebuild optimizer to include newly unfrozen params
        optimizer = torch.optim.Adam(
            filter(lambda p: p.requires_grad, model.parameters()),
            lr=scheduler.get_lr(), weight_decay=1e-4)
        scheduler.optimizer = optimizer
        backbone_unfrozen = True
        print(f"\nâ”€â”€ Epoch {epoch}: Backbone unfrozen â”€â”€")

    # Train
    model.train()
    tr_loss = tr_total = 0.0
    for frames, future, w in tqdm(train_dl,
            desc=f"Ep {epoch:>2} Train", ncols=80, leave=False):
        frames, future, w = frames.to(device), future.to(device), w.to(device)
        optimizer.zero_grad()
        pred = model(frames)
        loss = weighted_mse(pred, future, w)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        tr_loss  += loss.item() * frames.size(0)
        tr_total += frames.size(0)

    # Validate
    model.eval()
    va_loss = va_total = 0.0
    all_preds, all_futures, all_weights = [], [], []
    with torch.no_grad():
        for frames, future, w in tqdm(val_dl,
                desc=f"Ep {epoch:>2} Val  ", ncols=80, leave=False):
            frames, future, w = frames.to(device), future.to(device), w.to(device)
            pred     = model(frames)
            loss     = weighted_mse(pred, future, w)
            va_loss += loss.item() * frames.size(0)
            va_total+= frames.size(0)
            all_preds.append(pred.cpu().numpy())
            all_futures.append(future.cpu().numpy())
            all_weights.extend(w.cpu().numpy())

    tr_mse   = tr_loss / tr_total
    va_mse   = va_loss / va_total
    elapsed  = time.time() - t0

    pf = np.concatenate(all_preds).reshape(-1)
    ff = np.concatenate(all_futures).reshape(-1)
    pearson_r, _ = pearsonr(pf[:100000], ff[:100000])

    vs_persist = (persistence_mse - va_mse) / persistence_mse * 100
    vs_mean    = (mean_frame_mse  - va_mse) / mean_frame_mse  * 100
    new_lr, direction, rewound = scheduler.step(vs_mean, pearson_r)
    improved   = " âœ“" if va_mse < best_val_mse else ""

    history.append(dict(epoch=epoch, tr_mse=tr_mse, va_mse=va_mse,
                        pearson_r=pearson_r, lr=new_lr,
                        backbone_unfrozen=backbone_unfrozen))

    phase = "frozen" if not backbone_unfrozen else "fine-tune"
    print(f"Ep {epoch:>2}/{MAX_EPOCHS} [{phase}] | "
          f"tr={tr_mse:.5f} | va={va_mse:.5f} | "
          f"r={pearson_r:.4f} | "
          f"vs_persist={vs_persist:+.1f}% | "
          f"vs_mean={vs_mean:+.1f}% | "
          f"lr={new_lr:.2e} | {direction} | {elapsed:.0f}s{improved}")

    if va_mse < best_val_mse:
        best_val_mse = va_mse
        patience_ctr = 0
        torch.save(model.state_dict(), best_path)
    else:
        patience_ctr += 1
        if patience_ctr >= PATIENCE:
            print(f"\nEarly stopping at epoch {epoch}")
            break

print(f"\nâœ“ Best val MSE : {best_val_mse:.6f}")
print(f"  vs persistence : {(persistence_mse-best_val_mse)/persistence_mse*100:+.1f}%")
print(f"  vs mean-frame  : {(mean_frame_mse-best_val_mse)/mean_frame_mse*100:+.1f}%")

# â”€â”€ Final evaluation â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
model.load_state_dict(torch.load(best_path, map_location=device))
model.eval()
all_preds, all_futures, all_weights = [], [], []
with torch.no_grad():
    for frames, future, w in val_dl:
        pred = model(frames.to(device)).cpu().numpy()
        all_preds.append(pred)
        all_futures.append(future.numpy())
        all_weights.extend(w.numpy())

all_preds    = np.concatenate(all_preds,   axis=0)
all_futures  = np.concatenate(all_futures, axis=0)
all_weights  = np.array(all_weights)
per_snap_mse = ((all_preds - all_futures) ** 2).mean(axis=(1, 2))

val_current  = np.array([corr_video[s] for s in daily_val[:len(all_preds)]])
dir_acc      = (np.sign(all_preds - val_current) ==
                np.sign(all_futures - val_current)).mean()

pf = all_preds.reshape(-1);  ff = all_futures.reshape(-1)
pearson_r_final, _ = pearsonr(pf[:500000], ff[:500000])

volatile_mask = all_weights >= volatility_threshold
mse_volatile  = per_snap_mse[volatile_mask].mean() if volatile_mask.any() else float("nan")
mse_stable    = per_snap_mse[~volatile_mask].mean() if (~volatile_mask).any() else float("nan")

print(f"\nâ”€â”€ Final Evaluation â”€â”€")
print(f"  Val MSE (unweighted)  : {per_snap_mse.mean():.6f}  (CNN+LSTM: 0.072454)")
print(f"  Persistence MSE       : {persistence_mse:.6f}")
print(f"  Mean-frame MSE        : {mean_frame_mse:.6f}")
print(f"  Pearson r             : {pearson_r_final:.4f}  (CNN+LSTM: 0.2887)")
print(f"  Directional accuracy  : {dir_acc*100:.2f}%  (CNN+LSTM: 73.03%)")
print(f"  MSE on volatile snaps : {mse_volatile:.6f}  (CNN+LSTM: 0.109566)")
print(f"  MSE on stable snaps   : {mse_stable:.6f}  (CNN+LSTM: 0.036622)")

# â”€â”€ Save â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
np.save(os.path.join(PROJECT_DIR, "efficientnet_val_preds.npy"), all_preds)
with open(os.path.join(MODELS_DIR, "efficientnet_history.pkl"), "wb") as f:
    pickle.dump(history, f)
print(f"\nâœ“ Saved: {best_path}")

# â”€â”€ Release GPU â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
del model
torch.cuda.empty_cache()
print(f"âœ“ GPU memory released")

# â”€â”€ Plot â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
fig, axes = plt.subplots(1, 4, figsize=(22, 5))
fig.suptitle("EfficientNet-B0 + LSTM â€” Transfer Learning + Directional LR",
             fontsize=13, fontweight="bold")

epochs  = [h["epoch"]     for h in history]
tr_mses = [h["tr_mse"]    for h in history]
va_mses = [h["va_mse"]    for h in history]
rs      = [h["pearson_r"] for h in history]
lrs     = [h["lr"]        for h in history]

# Mark unfreeze epoch
unfreeze_ep = FREEZE_EPOCHS + 1

axes[0].plot(epochs, tr_mses, "o-", color="steelblue",  label="Train MSE")
axes[0].plot(epochs, va_mses, "o-", color="darkorange", label="Val MSE")
axes[0].axhline(persistence_mse, color="red",    lw=1.5,
                linestyle="--", label=f"Persistence {persistence_mse:.4f}")
axes[0].axhline(mean_frame_mse,  color="green",  lw=1.5,
                linestyle="--", label=f"Mean-frame {mean_frame_mse:.4f}")
axes[0].axhline(0.075715, color="purple", lw=1.2,
                linestyle=":", label="CNN+LSTM 0.0757")
axes[0].axvline(unfreeze_ep, color="gray", lw=1.5,
                linestyle="--", alpha=0.7, label="Backbone unfrozen")
axes[0].set_title("MSE over epochs"); axes[0].set_xlabel("Epoch")
axes[0].set_ylabel("MSE"); axes[0].legend(fontsize=7)
axes[0].grid(True, alpha=0.3)

axes[1].plot(epochs, rs, "o-", color="forestgreen")
axes[1].axhline(0.2887, color="purple", lw=1.2,
                linestyle=":", label="CNN+LSTM r=0.289")
axes[1].axvline(unfreeze_ep, color="gray", lw=1.5, linestyle="--", alpha=0.7)
axes[1].set_title("Pearson r"); axes[1].set_xlabel("Epoch")
axes[1].set_ylabel("Pearson r"); axes[1].set_ylim(0, 1)
axes[1].legend(fontsize=8); axes[1].grid(True, alpha=0.3)

axes[2].plot(epochs, lrs, "o-", color="crimson")
axes[2].axvline(unfreeze_ep, color="gray", lw=1.5,
                linestyle="--", alpha=0.7, label="Backbone unfrozen")
axes[2].set_title("Directional LR"); axes[2].set_xlabel("Epoch")
axes[2].set_ylabel("LR"); axes[2].set_yscale("log")
axes[2].legend(fontsize=8); axes[2].grid(True, alpha=0.3)

sample_idx = np.random.choice(min(len(pf), 500000), 5000, replace=False)
axes[3].scatter(ff[:500000][sample_idx], pf[:500000][sample_idx],
                alpha=0.1, s=1, color="steelblue")
axes[3].plot([-1,1], [-1,1], "r--", lw=1.5, label="Perfect")
axes[3].set_xlabel("Actual R_ewma"); axes[3].set_ylabel("Predicted R_ewma")
axes[3].set_title(f"Predicted vs Actual (r={pearson_r_final:.3f})")
axes[3].legend(fontsize=9); axes[3].grid(True, alpha=0.3)

plt.tight_layout()
path = os.path.join(PROJECT_DIR, "plot_cell16_efficientnet.png")
plt.savefig(path, dpi=150, bbox_inches="tight")
plt.show()
print(f"âœ“ Saved: {path}")
print("\nâœ“ Cell 16 complete. Run Cell 17 for ensemble.")
```

```python
# Evaluation Cell â€” Final Comparison Plots for Paper
#
# Loads all saved val predictions and produces publication-quality figures:
#   1. Model comparison bar chart (all metrics)
#   2. Per-snapshot MSE over time (volatile vs stable)
#   3. Predicted vs actual scatter â€” all models side by side
#   4. Directional accuracy breakdown (volatile vs stable)
#   5. Adaptive resolution ablation figure

import numpy as np
import pickle, os
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
from matplotlib.patches import Patch
from scipy.stats import pearsonr
from sklearn.metrics import mean_squared_error

PROJECT_DIR = r"C:\Users\HP\Documents\InterStockRelation_Paper\InterStockRelation"
MODELS_DIR  = os.path.join(PROJECT_DIR, "models")

plt.rcParams.update({
    "font.family"  : "serif",
    "font.size"    : 11,
    "axes.titlesize": 12,
    "axes.labelsize": 11,
    "figure.dpi"   : 150,
})

# â”€â”€ Load metadata â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
with open(os.path.join(PROJECT_DIR, "dataset_meta_v2.pkl"), "rb") as f:
    dm = pickle.load(f)
with open(os.path.join(PROJECT_DIR, "video_meta.pkl"), "rb") as f:
    vm = pickle.load(f)

corr_video           = np.load(os.path.join(PROJECT_DIR, "corr_video.npy"))
daily_val            = dm["daily_val"]
persistence_mse      = dm["persistence_mse"]
mean_frame_mse       = dm["mean_frame_mse"]
frame_mse            = dm["frame_mse"]
volatility_threshold = dm["volatility_threshold"]
snap_dates           = dm["snap_dates"]
HORIZON              = dm["horizon"]
N_STOCKS             = dm["n_stocks"]

# â”€â”€ Load all model predictions â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
print("Loading model predictions...")

pred_files = {
    "CNN+LSTM\n(daily)"    : None,   # daily-only not saved separately â€” skip
    "CNN+LSTM\n(adaptive)" : "cnn_lstm_val_preds.npy",
    "3D CNN\n(adaptive)"   : "3dcnn_val_preds.npy",
    "ConvLSTM\n(adaptive)" : "convlstm_val_preds.npy",
    "EfficientNet\n+LSTM"  : "efficientnet_val_preds.npy",
}

# Load available predictions
preds = {}
for name, fname in pred_files.items():
    if fname is None:
        continue
    fpath = os.path.join(PROJECT_DIR, fname)
    if os.path.exists(fpath):
        preds[name] = np.load(fpath)
        print(f"  âœ“ {name.replace(chr(10),' '):<30} shape={preds[name].shape}")
    else:
        print(f"  âœ— {name.replace(chr(10),' '):<30} NOT FOUND â€” skipping")

# Ground truth
n_val      = min(len(p) for p in preds.values())
val_snaps  = daily_val[:n_val]
futures    = np.array([corr_video[s + HORIZON] for s in val_snaps])
currents   = np.array([corr_video[s]           for s in val_snaps])
val_weights= frame_mse[val_snaps]
volatile_mask = val_weights >= volatility_threshold

print(f"\nVal snapshots : {n_val}")
print(f"Volatile      : {volatile_mask.sum()}  ({volatile_mask.mean()*100:.1f}%)")
print(f"Stable        : {(~volatile_mask).sum()}  ({(~volatile_mask).mean()*100:.1f}%)")

# â”€â”€ Compute metrics for all models â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
def compute_metrics(pred, future, current, volatile_mask):
    per_snap_mse = ((pred - future) ** 2).mean(axis=(1, 2))
    mse          = per_snap_mse.mean()
    mse_vol      = per_snap_mse[volatile_mask].mean()
    mse_sta      = per_snap_mse[~volatile_mask].mean()
    pf = pred.reshape(-1);  ff = future.reshape(-1)
    r, _         = pearsonr(pf[:500000], ff[:500000])
    dir_acc      = (np.sign(pred - current) ==
                    np.sign(future - current)).mean()
    vs_persist   = (persistence_mse - mse) / persistence_mse * 100
    vs_mean      = (mean_frame_mse  - mse) / mean_frame_mse  * 100
    return dict(mse=mse, mse_vol=mse_vol, mse_sta=mse_sta,
                pearson_r=r, dir_acc=dir_acc*100,
                vs_persist=vs_persist, vs_mean=vs_mean,
                per_snap_mse=per_snap_mse)

# Baselines
baseline_metrics = {
    "Persistence"  : dict(mse=persistence_mse,
                          mse_vol=np.nan, mse_sta=np.nan,
                          pearson_r=np.nan, dir_acc=50.0,
                          vs_persist=0.0, vs_mean=-88.8,
                          per_snap_mse=frame_mse[val_snaps]),
    "Mean-frame"   : dict(mse=mean_frame_mse,
                          mse_vol=np.nan, mse_sta=np.nan,
                          pearson_r=np.nan, dir_acc=np.nan,
                          vs_persist=47.0, vs_mean=0.0,
                          per_snap_mse=np.array([
                              np.mean((dm["train_mean_frame"] - futures[i])**2)
                              for i in range(n_val)])),
}

model_metrics = {}
for name, pred in preds.items():
    pred_aligned = pred[:n_val]
    model_metrics[name] = compute_metrics(
        pred_aligned, futures, currents, volatile_mask)

all_metrics = {**baseline_metrics, **model_metrics}

# Print summary table
print(f"\n{'Model':<28} {'MSE':>8} {'vs Pers':>8} {'vs Mean':>8} "
      f"{'r':>7} {'DirAcc':>8} {'MSE Vol':>9} {'MSE Sta':>9}")
print("-" * 92)
for name, m in all_metrics.items():
    def fmt(v, fmt_str):
        return ("N/A" if (isinstance(v, float) and np.isnan(v))
                else format(v, fmt_str))
    print(f"{name.replace(chr(10),' '):<28} "
          f"{fmt(m['mse'],        '8.5f')} "
          f"{fmt(m['vs_persist'], '+7.1f')}% "
          f"{fmt(m['vs_mean'],    '+7.1f')}% "
          f"{fmt(m['pearson_r'],  '7.4f')} "
          f"{fmt(m['dir_acc'],    '7.1f')}% "
          f"{fmt(m['mse_vol'],    '9.6f')} "
          f"{fmt(m['mse_sta'],    '9.6f')}")

# â”€â”€ Figure 1: Main comparison bar chart â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
model_names  = list(model_metrics.keys())
colors       = ["#2166ac", "#4dac26", "#d01c8b", "#f1a340"]
colors       = colors[:len(model_names)]

fig1, axes = plt.subplots(1, 4, figsize=(18, 5))
fig1.suptitle("Model Comparison â€” Future Correlation Matrix Prediction\n"
              "Nifty 100, 5-day horizon, stratified volatility validation set",
              fontsize=13, fontweight="bold")

# MSE
ax = axes[0]
vals = [model_metrics[n]["mse"] for n in model_names]
bars = ax.bar(range(len(model_names)), vals, color=colors, alpha=0.85,
              edgecolor="white", width=0.6)
ax.axhline(persistence_mse, color="red",   lw=1.5, linestyle="--",
           label=f"Persistence {persistence_mse:.4f}")
ax.axhline(mean_frame_mse,  color="green", lw=1.5, linestyle="--",
           label=f"Mean-frame {mean_frame_mse:.4f}")
ax.set_xticks(range(len(model_names)))
ax.set_xticklabels([n.replace("\n", "\n") for n in model_names],
                   fontsize=9)
ax.set_ylabel("MSE â†“"); ax.set_title("Validation MSE")
for bar, v in zip(bars, vals):
    ax.text(bar.get_x()+bar.get_width()/2, v+0.0005,
            f"{v:.4f}", ha="center", va="bottom", fontsize=8)
ax.legend(fontsize=8); ax.grid(True, alpha=0.3, axis="y")
ax.set_ylim(0.065, 0.145)

# Pearson r
ax = axes[1]
vals = [model_metrics[n]["pearson_r"] for n in model_names]
bars = ax.bar(range(len(model_names)), vals, color=colors, alpha=0.85,
              edgecolor="white", width=0.6)
ax.set_xticks(range(len(model_names)))
ax.set_xticklabels([n for n in model_names], fontsize=9)
ax.set_ylabel("Pearson r â†‘"); ax.set_title("Structural Agreement (Pearson r)")
for bar, v in zip(bars, vals):
    ax.text(bar.get_x()+bar.get_width()/2, v+0.002,
            f"{v:.3f}", ha="center", va="bottom", fontsize=8)
ax.grid(True, alpha=0.3, axis="y"); ax.set_ylim(0, 0.4)

# Directional accuracy
ax = axes[2]
vals = [model_metrics[n]["dir_acc"] for n in model_names]
bars = ax.bar(range(len(model_names)), vals, color=colors, alpha=0.85,
              edgecolor="white", width=0.6)
ax.axhline(50, color="red", lw=1.5, linestyle="--", label="Random 50%")
ax.set_xticks(range(len(model_names)))
ax.set_xticklabels([n for n in model_names], fontsize=9)
ax.set_ylabel("Accuracy (%) â†‘"); ax.set_title("Directional Accuracy")
for bar, v in zip(bars, vals):
    ax.text(bar.get_x()+bar.get_width()/2, v+0.1,
            f"{v:.1f}%", ha="center", va="bottom", fontsize=8)
ax.legend(fontsize=8); ax.grid(True, alpha=0.3, axis="y")
ax.set_ylim(60, 80)

# Volatile vs stable MSE
ax = axes[3]
x  = np.arange(len(model_names))
w  = 0.35
vol_vals = [model_metrics[n]["mse_vol"] for n in model_names]
sta_vals = [model_metrics[n]["mse_sta"] for n in model_names]
ax.bar(x - w/2, vol_vals, w, color="crimson",    alpha=0.8,
       edgecolor="white", label="Volatile")
ax.bar(x + w/2, sta_vals, w, color="steelblue",  alpha=0.8,
       edgecolor="white", label="Stable")
ax.set_xticks(x)
ax.set_xticklabels([n for n in model_names], fontsize=9)
ax.set_ylabel("MSE â†“"); ax.set_title("MSE: Volatile vs Stable Snapshots")
ax.legend(fontsize=9); ax.grid(True, alpha=0.3, axis="y")

plt.tight_layout()
path1 = os.path.join(PROJECT_DIR, "plot_eval_model_comparison.png")
plt.savefig(path1, dpi=150, bbox_inches="tight")
plt.show()
print(f"âœ“ Saved: {path1}")

# â”€â”€ Figure 2: Per-snapshot MSE over time â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
fig2, axes2 = plt.subplots(len(model_metrics), 1,
                            figsize=(14, 3*len(model_metrics)),
                            sharex=True)
if len(model_metrics) == 1:
    axes2 = [axes2]

fig2.suptitle("Per-Snapshot MSE Over Time â€” All Models\n"
              "(orange = volatile snapshot, blue = stable)",
              fontsize=12, fontweight="bold")

val_dates = [snap_dates[s] for s in val_snaps]

for ax, (name, m) in zip(axes2, model_metrics.items()):
    psmse = m["per_snap_mse"]
    for i, (d, v, vol) in enumerate(zip(val_dates, psmse, volatile_mask)):
        color = "darkorange" if vol else "steelblue"
        ax.axvline(d, color=color, alpha=0.4, lw=1.5)
    ax.plot(val_dates, psmse, lw=0.8, color="black", alpha=0.7, zorder=3)
    ax.axhline(persistence_mse, color="red",   lw=1.2, linestyle="--",
               alpha=0.7, label=f"Persistence {persistence_mse:.4f}")
    ax.axhline(mean_frame_mse,  color="green", lw=1.2, linestyle="--",
               alpha=0.7, label=f"Mean-frame {mean_frame_mse:.4f}")
    ax.set_ylabel("MSE"); ax.set_title(name.replace("\n"," "))
    ax.legend(fontsize=7, loc="upper right"); ax.grid(True, alpha=0.2)

import matplotlib.dates as mdates
axes2[-1].xaxis.set_major_formatter(mdates.DateFormatter("%Y-%m"))
axes2[-1].xaxis.set_major_locator(mdates.MonthLocator(interval=3))
plt.setp(axes2[-1].xaxis.get_majorticklabels(), rotation=30, ha="right")

legend_elements = [
    Patch(color="darkorange", alpha=0.6, label="Volatile snapshot"),
    Patch(color="steelblue",  alpha=0.6, label="Stable snapshot"),
]
axes2[0].legend(handles=legend_elements, fontsize=8, loc="upper left")

plt.tight_layout()
path2 = os.path.join(PROJECT_DIR, "plot_eval_per_snap_mse.png")
plt.savefig(path2, dpi=150, bbox_inches="tight")
plt.show()
print(f"âœ“ Saved: {path2}")

# â”€â”€ Figure 3: Predicted vs actual scatter â€” all models â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
n_models = len(model_metrics)
fig3, axes3 = plt.subplots(1, n_models, figsize=(5*n_models, 5))
if n_models == 1:
    axes3 = [axes3]

fig3.suptitle("Predicted vs Actual Correlation Values â€” All Models\n"
              "Validation set (5000 randomly sampled pair-snapshots)",
              fontsize=12, fontweight="bold")

for ax, (name, m), color in zip(axes3, model_metrics.items(), colors):
    pred = preds[name][:n_val].reshape(-1)
    fut  = futures.reshape(-1)
    idx  = np.random.choice(min(len(pred), 500000), 5000, replace=False)
    r    = m["pearson_r"]
    ax.scatter(fut[idx], pred[idx], alpha=0.1, s=1, color=color)
    ax.plot([-1,1], [-1,1], "r--", lw=1.5, label="Perfect")
    ax.set_xlabel("Actual R_ewma")
    ax.set_ylabel("Predicted R_ewma")
    ax.set_title(f"{name.replace(chr(10),' ')}\n(r={r:.3f})")
    ax.set_xlim(-1.1, 1.1); ax.set_ylim(-1.1, 1.1)
    ax.legend(fontsize=8); ax.grid(True, alpha=0.3)
    ax.set_aspect("equal")

plt.tight_layout()
path3 = os.path.join(PROJECT_DIR, "plot_eval_scatter_all.png")
plt.savefig(path3, dpi=150, bbox_inches="tight")
plt.show()
print(f"âœ“ Saved: {path3}")

# â”€â”€ Figure 4: Trial 1 vs Trial 2 comparison â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
# Shows the ceiling hit in Trial 1 (pair-level) vs Trial 2 (matrix-level)
fig4, ax4 = plt.subplots(figsize=(12, 5))
fig4.suptitle("Trial 1 (Pair-Level Classification) vs Trial 2 (Matrix Regression)\n"
              "Demonstrating the ceiling of pair-level approaches",
              fontsize=12, fontweight="bold")

trial1_models  = ["Random\nBaseline", "XGBoost", "LSTM", "Ensemble\nXGB+LSTM",
                  "TransformerGNN"]
trial1_acc     = [33.3, 55.6, 55.8, 56.1, 55.5]
trial1_colors  = ["lightgray"] + ["#d1e5f0"]*4

trial2_dir_acc = [model_metrics[n]["dir_acc"] for n in model_metrics]
trial2_names   = [n.replace("\n"," ") for n in model_metrics]
trial2_colors  = colors

x1 = np.arange(len(trial1_models))
x2 = np.arange(len(trial2_names)) + len(trial1_models) + 1

bars1 = ax4.bar(x1, trial1_acc,     color=trial1_colors, edgecolor="gray",
                width=0.6, alpha=0.85, label="Trial 1 (classification accuracy)")
bars2 = ax4.bar(x2, trial2_dir_acc, color=trial2_colors, edgecolor="white",
                width=0.6, alpha=0.85, label="Trial 2 (directional accuracy)")

ax4.axhline(33.3, color="gray",   lw=1.2, linestyle=":",
            label="Random (3-class) 33.3%")
ax4.axhline(50.0, color="orange", lw=1.2, linestyle=":",
            label="Random (binary) 50.0%")
ax4.axvline(len(trial1_models) + 0.5, color="black", lw=2,
            linestyle="--", alpha=0.5)
ax4.text(len(trial1_models)/2 - 0.5, 58, "Trial 1\n(Pair-level\nclassification)",
         ha="center", fontsize=10, color="gray")
ax4.text(len(trial1_models) + 1 + len(trial2_names)/2, 77,
         "Trial 2\n(Matrix-level\nregression)",
         ha="center", fontsize=10, color="steelblue")

all_x     = list(x1) + list(x2)
all_names = trial1_models + trial2_names
ax4.set_xticks(all_x)
ax4.set_xticklabels(all_names, fontsize=9, rotation=15, ha="right")
ax4.set_ylabel("Accuracy (%)")
ax4.set_ylim(0, 85)
ax4.legend(fontsize=9, loc="upper left")
ax4.grid(True, alpha=0.3, axis="y")

for bar, v in zip(bars1, trial1_acc):
    ax4.text(bar.get_x()+bar.get_width()/2, v+0.5,
             f"{v:.1f}%", ha="center", va="bottom", fontsize=8)
for bar, v in zip(bars2, trial2_dir_acc):
    ax4.text(bar.get_x()+bar.get_width()/2, v+0.5,
             f"{v:.1f}%", ha="center", va="bottom", fontsize=8)

plt.tight_layout()
path4 = os.path.join(PROJECT_DIR, "plot_eval_trial1_vs_trial2.png")
plt.savefig(path4, dpi=150, bbox_inches="tight")
plt.show()
print(f"âœ“ Saved: {path4}")

print(f"\nâ”€â”€ All evaluation plots saved â”€â”€")
print(f"  {path1}")
print(f"  {path2}")
print(f"  {path3}")
print(f"  {path4}")
print(f"\nAdd these to the paper LaTeX as \\includegraphics figures.")
```
